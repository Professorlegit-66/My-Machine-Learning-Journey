# Exploratory Data Analysis (EDA):
- We can consider every column a variable
- An independent analysis of a single variable (1 column) is known as univariant analysis
- An analysis of two variables (2 columns) at the same time is known as unibariant analysis
- An analysis of three variables (3 columns) at the same time is known as univariant analysis

# 1: UniVariate Analysis:
# Types of Data:
- Numerical Data
- Categorical Data

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/train.csv')
df.head()
```

# 1: Categorical Data
a: Countplot


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(df, x='Survived')
```

```python
sns.countplot(df, x='Pclass')
```

```python
sns.countplot(df, x='Embarked')
```

b: PieChart

```python
df['Pclass'].value_counts().plot(kind='pie',autopct='%.2f')
```

# 2: Numerical Data
a: Histogram
- To understand the distribution of data in that numerical column

```python
plt.hist(df['Age'],bins=15)
```

b: DistPlot
- Tell us the probability of a number occuring
- Tell us about the skewness of a data which can be left skewed or right skewed

```python
sns.distplot(df['Age'])
```

```python
df['Age'].skew()
```

c: BoxPlot
- Gives us the 5 number summary

```python
sns.boxplot(df['Age'],orient='y')
```
