```python
import pandas as pd
import seaborn as sns
```

# 2: BiVariant Analysis:

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
df = pd.read_csv('train.csv')
df.head()
```

```python
flights = sns.load_dataset('flights')
flights.head()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

# 1: Scatter Plot (Numerical - Numerical):

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

# 2: Bar Plot (Numerical - Categorical):

```python
df.head()
```

X-Axis generally has Categorical Data while Y-Axis generally has Numerical Data

```python
sns.barplot(data=df,x='Pclass',y='Age')
```

```python
sns.barplot(data=df,x='Pclass',y='Fare')
```

# 3: Box Plot (Numerical - Categorical):

```python
sns.boxplot(data=df,x='Sex',y='Age')
```

# 4: Dist Plot (Numerical - Categorical):

```python
sns.distplot(df[df['Survived'] == 0]['Age'],hist=False)
sns.distplot(df[df['Survived'] == 1]['Age'],hist=False)
```

# 5: Heat Map (Categorical - Categorical):

```python
df.head()
```

```python
pd.crosstab(df['Pclass'],df['Survived'])
```

```python
sns.heatmap(pd.crosstab(df['Pclass'],df['Survived']))
```

```python
(df.groupby('Sex').mean(numeric_only=True)['Survived']*100)
```

```python
(df.groupby('Sex').mean(numeric_only=True)['Survived']*100).plot(kind='bar')
```

```python
(df.groupby('Embarked').mean(numeric_only=True)['Survived']*100)
```

# 6: Cluster Map (Categorical - Categorical):

```python
pd.crosstab(df['SibSp'],df['Survived'])
```

```python
sns.clustermap(pd.crosstab(df['SibSp'],df['Survived']))
```
