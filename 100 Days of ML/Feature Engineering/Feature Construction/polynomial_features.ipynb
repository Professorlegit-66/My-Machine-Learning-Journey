```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/train.csv').drop(columns=['PassengerId','Survived','SibSp','Parch','Pclass','Name','Sex','Ticket','Embarked','Cabin'])
df.head(5)
```

## 5. Polynomial Features:
Using `Age` and `Fare` as the base numeric columns.

```python
poly_df = df[['Age', 'Fare']].dropna()
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(poly_df)
```

```python
poly_feature_names = poly.get_feature_names_out(['Age', 'Fare'])
```

```python
poly_result = pd.DataFrame(poly_features, columns=poly_feature_names, index=poly_df.index)
poly_result.head()
```
