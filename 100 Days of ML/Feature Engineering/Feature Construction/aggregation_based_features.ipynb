```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/train.csv').drop(columns=['Sex','Age','Ticket','Cabin','Embarked'])
df.head(5)
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df[['SibSp', 'Parch', 'FamilySize']].head()
```

## 4. Aggregation-Based (Grouped) Features:

### a. Average Fare per Passenger Class:

```python
class_fare_avg = df.groupby('Pclass')['Fare'].transform('mean')
df['AvgFareByClass'] = class_fare_avg
df[['Pclass', 'Fare', 'AvgFareByClass']].head()
```

### b. Family Group Size by Surname:

```python
df['Title'] = df['Name'].str.extract(r',\s*([^\.]*)\.')
df['Title'].value_counts()
```

```python
df['Surname'] = df['Name'].str.extract(r'^([^,]*),')
df[['Name', 'Surname', 'Title']].head()
```

```python
surname_counts = df.groupby('Surname')['PassengerId'].transform('count')
df['SurnameGroupSize'] = surname_counts
df[['Surname', 'FamilySize', 'SurnameGroupSize']].head(10)
```

### c. Survival Rate by Title:

```python
title_survival_rate = df.groupby('Title')['Survived'].transform('mean')
df['TitleSurvivalRate'] = title_survival_rate
df[['Title', 'Survived', 'TitleSurvivalRate']].head(10)
```
