```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/train.csv').drop(columns=['PassengerId','Pclass','Name','Sex','Ticket','Embarked'])
df.head(5)
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df[['SibSp', 'Parch', 'FamilySize']].head()
```

## 3. Binary / Flag Features:

### a. Is Alone:

```python
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
df[['FamilySize', 'IsAlone']].head()
```

### b. Is Child:

```python
df['IsChild'] = (df['Age'] < 18).astype(int)
df[['Age', 'IsChild']].head()
```

### c. Has Cabin Recorded:

```python
df['HasCabin'] = df['Cabin'].notna().astype(int)
df[['Cabin', 'HasCabin']].head()
```

### d. Is High Fare:

```python
fare_75th = df['Fare'].quantile(0.75)
df['IsHighFare'] = (df['Fare'] > fare_75th).astype(int)
df[['Fare', 'IsHighFare']].head()
```
