```python
import pandas as pd
import numpy as np

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
df.head(5)
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X.head()
```

```python
y.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```

## Feature Construction:

```python
X['family_size'] = X['SibSp'] + X['Parch'] + 1
X.head()
```

```python
def size(num):
    if num == 1:
        return 0
    elif num > 1 and num <= 4:
        return 1
    else:
        return 2
```

```python
X['family_type'] = X['family_size'].apply(size)
```

```python
X.sample(10)
```

```python
X.drop(columns=['SibSp','Parch','family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```
