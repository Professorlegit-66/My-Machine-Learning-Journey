# Feature Engineering: The Four Categories

Before diving in, here's the umbrella context — feature engineering is generally split into four categories:

| Category | Purpose |
|---|---|
| **Feature Transformation** | Change the *scale/distribution/representation* of existing features (values change, count of features usually stays the same) |
| **Feature Construction/Extraction** | *Create new* features from existing ones (e.g., PCA components, combining columns) |
| **Feature Selection** | *Choose a subset* of existing features, discarding the rest (no new values created) |
| **Feature Extraction** | Reduce dimensionality by deriving new compressed representations (e.g., PCA, LDA, autoencoders) |

An outlier is a data point that deviates significantly from the overall pattern of a dataset — it lies unusually far from other observations in a way that raises suspicion it was generated by a different mechanism than the rest of the data.

A commonly cited formal definition, from statistician Douglas Hawkins, is:

#### C: Outlier detection:<bx>
> "An outlier is an observation that deviates so much from other observations as to arouse suspicion that it was generated by a different mechanism."

**Outlier detection**, then, is the formal task of identifying such observations in a dataset — i.e., finding data points that are statistically or contextually inconsistent with the rest of the data. It's often framed as an unsupervised learning problem, since you usually don't have labeled "outlier" vs "normal" examples in advance.

A few important nuances worth knowing:

1. **It's not just "extreme values."** An outlier is defined relative to the underlying data-generating process, not just numeric magnitude. A value can be extreme but not an outlier (if it's a legitimate rare event), or non-extreme but still an outlier (in multivariate contexts, a point can be an outlier due to an unusual *combination* of feature values even if each individual feature looks normal).

2. **Common categories:**
   - **Point/Global outliers** — a single point that's anomalous relative to the entire dataset (e.g., one transaction of $1M among normal $50 transactions).
   - **Contextual outliers** — anomalous only in a specific context (e.g., 35°C is normal in summer but an outlier in winter).
   - **Collective outliers** — a group of points that's anomalous together, even if individual points aren't (e.g., a sudden sustained spike in network traffic).

3. **Common detection approaches:**
   - Statistical: Z-score, IQR (interquartile range) method
   - Distance/density-based: k-NN, DBSCAN, LOF (Local Outlier Factor)
   - Model-based: Isolation Forest, One-Class SVM
   - Visualization-based: box plots, scatter plots

### D: Feature Scaling:<br>
**Feature scaling** is the process of transforming the numeric features of a dataset so that they are on a common scale, without distorting the relative differences in the values or the underlying distribution shape — typically so that no single feature dominates a model simply because of the units or range it happens to be measured in.

Formally, it's a normalization step applied during preprocessing to bring all features into a comparable range or distribution, since many ML algorithms are sensitive to the magnitude of input values rather than just their relative relationships.

**Why it matters:**
Algorithms that rely on distance calculations, gradients, or regularization are especially sensitive to feature scale:
- **Distance-based**: KNN, K-Means, SVM — a feature ranging 0–100,000 (e.g., salary) will dominate a feature ranging 0–1 (e.g., a ratio) purely due to scale, not actual importance.
- **Gradient descent–based**: Linear/Logistic Regression, Neural Networks — unscaled features cause uneven, elongated loss surfaces, slowing or destabilizing convergence.
- **Regularized models**: Ridge/Lasso — penalties are applied to coefficient magnitudes, which are scale-dependent.

Tree-based models (Decision Trees, Random Forests, XGBoost) are generally scale-invariant, since they split on thresholds rather than distances — so scaling is often skipped for those.

**Common techniques:**

| Method | Formula | Effect |
|---|---|---|
| **Min-Max Scaling (Normalization)** | (x − min) / (max − min) | Rescales to a fixed range, usually [0, 1] |
| **Standardization (Z-score)** | (x − μ) / σ | Centers data at mean 0, std dev 1 |
| **Robust Scaling** | (x − median) / IQR | Uses median/IQR instead of mean/std — resistant to outliers |
| **MaxAbs Scaling** | x / max(\|x\|) | Scales to [-1, 1], preserves sparsity (good for sparse data) |

A key nuance: scaling should be **fit only on training data**, then applied (transformed) to validation/test data using those same fitted parameters — fitting on the full dataset causes data leakage.

Since this connects to your outlier discussion — Robust Scaling is often preferred over Standardization when outliers are present, since mean and std dev are themselves sensitive to outliers, while median and IQR aren't.

## Standardization (Z-score Scaling)

Transforms data to have **mean = 0** and **standard deviation = 1**, without bounding it to a fixed range.

$$x' = \frac{x - \mu}{\sigma}$$

- μ = mean of the feature, σ = standard deviation
- Resulting values are typically roughly in the range −3 to +3 (but not bounded)
- Preserves the shape of the original distribution — doesn't force it into a specific range
- Doesn't get distorted much by moderate outliers, but is still affected since mean/std are used
- Best for algorithms assuming/benefiting from zero-centered data: Linear/Logistic Regression, SVM, PCA, Neural Networks

**Use when:** the data roughly follows a Gaussian-like distribution, or the algorithm assumes centered data (e.g., PCA, which is variance-based).

---

## Normalization (Min-Max Scaling)

Rescales data into a **fixed, bounded range**, usually [0, 1].

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

- Bounded output — always between 0 and 1 (or whatever range you set)
- Very sensitive to outliers — one extreme value compresses everything else into a tiny sub-range
- Doesn't preserve the original distribution shape as faithfully as standardization

**Use when:** you know the approximate bounds of your data and there are few/no outliers — common for image pixel values (0–255 → 0–1), or neural network inputs where bounded ranges help activation functions.

---

## Important Types of Normalization

| Type | Formula | Notes |
|---|---|---|
| **Min-Max Normalization** | (x − min) / (max − min) | Standard [0,1] scaling; sensitive to outliers |
| **Mean Normalization** | (x − mean) / (max − min) | Centers around 0, but still range-bounded; less common |
| **Max Abs Scaling** | x / \|max\| | Scales to [−1, 1]; preserves zero entries → good for **sparse data** |
| **Robust Scaling** | (x − median) / IQR | Uses median/IQR instead of min/max/mean → resistant to outliers |
| **Unit Vector Normalization (L1/L2 norm)** | x / ‖x‖ | Scales each *row* (sample) to unit norm, not each column/feature — used in text data (TF-IDF), KNN, cosine similarity contexts |
| **Log Transformation** | log(x + 1) | Not strictly "normalization" but often paired with it — compresses skewed/heavy-tailed distributions before scaling |

---

## Standardization vs. Normalization — Quick Comparison

| Aspect | Standardization | Normalization |
|---|---|---|
| Output range | Unbounded (~ −3 to +3 typically) | Bounded (e.g., [0,1]) |
| Sensitive to outliers? | Somewhat | Very |
| Preserves distribution shape? | Yes | Less so |
| Best for | Gaussian-ish data, distance/gradient-based models | Bounded-input needs, e.g., image data, neural nets |
| Common use case | Regression, SVM, PCA, clustering | Image processing, deep learning inputs |

**Rule of thumb:** if you're unsure and outliers are present or the distribution isn't Gaussian, standardization (or Robust Scaling if outliers are heavy) is usually the safer default over Min-Max Normalization.

A key point: scaling should be fit only on training data, then applied (transformed) to validation/test data using those same fitted parameters — fitting on the full dataset causes data leakage.

# 1. Column Transformer

**Category: Not a transformer itself — a utility/orchestration tool for applying transformations**

`ColumnTransformer` (from `sklearn.compose`) isn't a mathematical transformation — it's a **container** that lets you apply *different* transformers to *different columns* of your dataset in a single step.

**Why it's needed:** Real datasets are mixed — numeric columns need scaling, categorical columns need encoding, text columns need vectorizing. Without `ColumnTransformer`, you'd have to manually split the dataframe, transform each part, then concatenate — messy and error-prone (especially about column order and train/test consistency).

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

ct = ColumnTransformer(transformers=[
    ('num', StandardScaler(), ['age', 'salary']),
    ('cat', OneHotEncoder(), ['city', 'gender'])], remainder='passthrough')  # or 'drop'

X_transformed = ct.fit_transform(X)
```

- `remainder='passthrough'` keeps untouched columns as-is; `'drop'` removes them.
- Output is a single combined array/matrix — ready for modeling.
- It's the standard way to feed mixed-type data into a **Pipeline** (see below).

---

# 2. Function Transformer

**Category: Feature Transformation**

`FunctionTransformer` (from `sklearn.preprocessing`) lets you wrap **any custom Python function** (including NumPy functions) into a scikit-learn–compatible transformer, so it can slot into a `Pipeline` or `ColumnTransformer` just like `StandardScaler` or `OneHotEncoder`.

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np

log_transformer = FunctionTransformer(np.log1p, validate=True)
X_transformed = log_transformer.fit_transform(X)
```

- `np.log1p` = log(1+x), safer than `np.log` for zero values.
- You can pass any function — custom-defined, lambda, or NumPy/SciPy built-in.
- Doesn't "learn" anything from the data (no `fit` parameters stored) — it just applies the function on `transform`. This is what distinguishes it from something like `StandardScaler`, which learns mean/std during `fit`.
- Useful for applying Log, Reciprocal, Square Root transforms (below) inside a pipeline in a clean, reusable way.

---

# 3. Log Transformer

**Category: Feature Transformation**

$$x' = \log(x) \quad \text{or commonly} \quad x' = \log(1 + x)$$

- Compresses the range of large values, expands the range of small values — reduces right-skewness (long right tail).
- Turns multiplicative relationships into additive ones — useful when a feature's *effect* is proportional/percentage-based rather than absolute (e.g., income, population, price).
- **Constraint:** undefined for x ≤ 0. Use `log1p` (log(1+x)) to safely handle zeros; for negative values, you'd need to shift the data first or use Yeo-Johnson instead.
- Common use: right-skewed features like income, house prices, word counts.

---

# 4. Reciprocal Transformer

**Category: Feature Transformation**

$$x' = \frac{1}{x}$$

- Inverts the relationship — large values become small, small values become large.
- Strongly reduces the influence of large outliers (since 1/large_number ≈ 0), but is extremely sensitive near x = 0 (blows up toward infinity).
- **Constraint:** undefined at x = 0; changes the *interpretation* of the relationship (not just scale) — so use only when the inverse relationship genuinely makes domain sense (e.g., converting "time to complete" into "rate," density into specific volume).
- Less commonly used than Log or Box-Cox because of its instability near zero and the fact it flips ordering of the variable's relationship with target in a way that's harder to interpret.

---

# 5. Square Root Transformer

**Category: Feature Transformation**

$$x' = \sqrt{x}$$

- Milder than Log — compresses large values less aggressively.
- Works only for **x ≥ 0** (undefined for negative numbers without adjustment).
- Often used for **moderately** right-skewed data (e.g., count data — number of children, number of events) where Log would over-correct.
- Special case of the **Power Transform** family, where the power = 0.5.

---

# 6. Power Transformer

**Category: Feature Transformation**

A general family:

$$x' = x^{\lambda}$$

Where λ (lambda) is a chosen or optimized exponent. Special cases:
- λ = 1 → no transformation
- λ = 0.5 → square root
- λ = -1 → reciprocal
- λ → 0 → approaches log transform (limiting case)

Rather than manually picking λ, scikit-learn's `PowerTransformer` class **automatically finds the optimal λ** for making the data as close to a **Gaussian (normal) distribution** as possible, using either:
1. **Box-Cox** method
2. **Yeo-Johnson** method

```python
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method='yeo-johnson')  # or 'box-cox'
X_transformed = pt.fit_transform(X)
```

So "Power Transformer" in scikit-learn is really the umbrella tool that implements Box-Cox and Yeo-Johnson under the hood — both of which are described next.

---

# 7. Box-Cox Transformer

**Category: Feature Transformation**

A specific, statistically optimized power transform:

$$
x' =
\begin{cases}
\dfrac{x^{\lambda} - 1}{\lambda}, & \lambda \neq 0 \\[6pt]
\ln(x), & \lambda = 0
\end{cases}
$$

- λ is estimated automatically (via maximum likelihood) to make the transformed data **as close to normally distributed as possible**.
- **Strict constraint: works only on strictly positive data (x > 0)** — this is its biggest limitation. Zero or negative values will throw an error, requiring a manual shift (e.g., add a constant) before applying it.
- Named after statisticians George Box and David Cox (1964).
- Very effective for stabilizing variance and normalizing skewed distributions — often used before applying models that assume normality (e.g., Linear Regression residual assumptions).

---

# 8. Yeo-Johnson Transformer

**Category: Feature Transformation**

An extension of Box-Cox designed to overcome its positivity constraint. Handles **zero and negative values** too:

$$
x' =
\begin{cases}
\dfrac{(x+1)^{\lambda} - 1}{\lambda}, & \lambda \neq 0, \, x \geq 0 \\[6pt]
\ln(x+1), & \lambda = 0, \, x \geq 0 \\[6pt]
-\dfrac{(-x+1)^{2-\lambda} - 1}{2-\lambda}, & \lambda \neq 2, \, x < 0 \\[6pt]
-\ln(-x+1), & \lambda = 2, \, x < 0
\end{cases}
$$

- Works for **any real-valued data** — no positivity constraint, unlike Box-Cox.
- This is why `PowerTransformer(method='yeo-johnson')` is scikit-learn's **default** — it's more broadly applicable.
- Reduces to something equivalent to Box-Cox behavior when all values are positive.
- Named after Yeo & Johnson (2000), who proposed it specifically to generalize Box-Cox to the full real line.

**Box-Cox vs. Yeo-Johnson — quick comparison:**

| Aspect | Box-Cox | Yeo-Johnson |
|---|---|---|
| Handles negative/zero values | ❌ No (x > 0 only) | ✅ Yes |
| Goal | Normalize distribution | Normalize distribution |
| Default in sklearn `PowerTransformer` | No | ✅ Yes |
| Flexibility | Less flexible | More flexible |

# Discretization (Binning)

**Category: Feature Transformation**

Discretization is the process of converting a **continuous numeric variable** into a **discrete/categorical variable** by grouping values into a finite number of intervals (bins). Each bin represents a range of values, and every original value is replaced by the bin/interval it falls into.

**Why use discretization?**
- To handle outliers (extreme values just get absorbed into the edge bins instead of skewing the model)
- To improve the value spread / handle skewed distributions
- To make relationships more robust to noise — small fluctuations in raw values no longer change the bin
- Some models (like linear models) can capture non-linear patterns better once continuous variables are bucketed

---

## I. Unsupervised Binning

These methods don't use the target variable (`y`) at all — bins are formed purely from the feature's own distribution.

### 1. Equal Width Binning (Uniform Binning)

Divides the range of the variable into **k bins of equal width**.

$$
\text{width} = \frac{x_{max} - x_{min}}{k}
$$

$$
\text{bin edges} = x_{min},\ x_{min}+\text{width},\ x_{min}+2\cdot\text{width},\ \dots,\ x_{max}
$$

- Each bin covers the same range of values, but can end up with **very different numbers of observations** per bin (if data is skewed, some bins may be nearly empty, others overloaded).
- Simple, fast, easy to interpret.
- Sensitive to outliers — a single extreme value stretches the range and makes most bins nearly empty.
- In sklearn: `KBinsDiscretizer(strategy='uniform')`

**Example:** Age 0–100 split into 5 equal-width bins → [0–20), [20–40), [40–60), [60–80), [80–100].

---

### 2. Equal Frequency Binning (Quantile Binning)

Divides the data so that **each bin contains (approximately) the same number of observations**, using quantiles as bin edges.

$$
\text{bin edges} = Q_0,\ Q_{1/k},\ Q_{2/k},\ \dots,\ Q_{k/k}
$$

Where each edge is a quantile of the data (e.g., for k=4 bins, edges are the 0th, 25th, 50th, 75th, 100th percentiles).

- Bin **widths vary**, but each bin has roughly equal population — good for skewed data.
- More robust to outliers than Equal Width, since it's based on rank/position, not raw magnitude.
- In sklearn: `KBinsDiscretizer(strategy='quantile')`

**Example:** Income data split into 4 equal-frequency bins (quartiles) → each bin holds 25% of the observations, regardless of the dollar range each bin spans.

---

### 3. K-Means Binning

Uses the **K-Means clustering algorithm** (1-dimensional, since it's applied to a single feature) to determine bin boundaries — cluster centers become the reference points, and values are assigned to bins based on which cluster centroid they're closest to.

$$
\text{minimize} \sum_{i=1}^{k} \sum_{x \in C_i} (x - \mu_i)^2
$$

Where μᵢ is the centroid of cluster/bin Cᵢ.

- Bin edges are placed based on the **natural density/clustering structure** of the data, rather than fixed width or fixed frequency.
- Good when the data has natural groupings/multi-modal distribution (e.g., clear clusters of low, medium, high values).
- More computationally expensive than the two methods above.
- In sklearn: `KBinsDiscretizer(strategy='kmeans')`

---

## II. Supervised Binning

Uses the **target variable (`y`)** to decide bin boundaries — bins are chosen to maximize the relationship between the binned feature and the target.

### 4. Decision Tree Binning

A **Decision Tree** is trained using the single continuous feature as the only input (X) and the target variable as the output (y). The **split points** the tree learns become the bin edges.

- Since decision trees split on thresholds that best separate the target classes (using Gini impurity or entropy for classification, variance reduction for regression), the resulting bins are **optimized to be predictive of the target** — not just descriptive of the feature's own distribution.
- Produces bins that directly improve model performance for the specific target being predicted.
- Risk: can overfit to the training data if the tree is too deep/unconstrained (should limit `max_depth` or `max_leaf_nodes`).
- Implementation: fit a shallow `DecisionTreeClassifier`/`DecisionTreeRegressor` on the single feature vs. target, then extract `tree_.threshold` values as bin edges.

---

## III. Custom Binning

Manually defined bin edges based on **domain knowledge** rather than any statistical or algorithmic method.

- You explicitly specify the boundaries yourself, e.g.:
  ```python
  bins = [0, 18, 35, 60, 100]
  labels = ['Minor', 'Young Adult', 'Adult', 'Senior']
  pd.cut(df['age'], bins=bins, labels=labels)
  ```
- Useful when there's established domain logic (e.g., age brackets for legal/marketing purposes, income tax brackets, BMI categories) that wouldn't naturally emerge from a purely statistical method.
- No formula — bin edges are chosen manually/heuristically based on subject-matter expertise.

---

## Summary Table

| Method | Uses Target? | Bin Basis | Sensitive to Outliers? |
|---|---|---|---|
| Equal Width | ❌ No | Fixed range width | ✅ Yes (very) |
| Equal Frequency | ❌ No | Quantiles (equal population) | ⚠️ Less sensitive |
| K-Means | ❌ No | Cluster centroids | ⚠️ Moderate |
| Decision Tree | ✅ Yes | Splits that best predict target | ⚠️ Moderate (depends on tree depth) |
| Custom | N/A | Domain knowledge | N/A (manual) |

All of these fall under **Feature Transformation** — they don't create entirely new information from scratch (that's Extraction) and they don't discard existing columns (that's Selection); they reshape how the existing continuous values are represented.

# Handling Missing Values

Missing values occur when no data is stored for a feature in a given observation. How you handle them can significantly affect model performance, so the choice of technique matters based on the type of data and *why* it's missing (MCAR, MAR, MNAR).

---

# A. Remove Them (Deletion)

The simplest approach — drop rows or columns with missing values.

- **Listwise deletion (row-wise):** Drop entire rows that contain any missing value (`df.dropna()`).
- **Column-wise deletion:** Drop the entire column if it has too high a percentage of missing values (e.g., >50-60%).

**When to use:** Only when missing data is a small fraction of the dataset (rule of thumb: <5%) and is **Missing Completely at Random (MCAR)** — i.e., missingness has no pattern/relationship with any other variable.

**Downside:** Loses information — can bias the dataset if missingness isn't random, and wastes data you can't afford to lose in smaller datasets.

# CCA (Complete Case Analysis)

CCA is simply the **formal statistical name** for what's shown in your notebook as "A: Remove them" — it's not a separate technique, it's the technical term for row-wise deletion of missing data.

---

## Definition

**Complete Case Analysis** means analyzing only the observations (rows) that have **complete data across all variables** — i.e., any row with even a single missing value in any column is dropped entirely, and only "complete cases" remain in the dataset.

```python
df_cca = df.dropna()  # this IS Complete Case Analysis
```

---

## When CCA is Appropriate

CCA relies on a key statistical assumption: the data must be **Missing Completely at Random (MCAR)**.

$$
P(\text{missing}) = \text{constant, independent of any variable (observed or unobserved)}
$$

- If missingness has **no relationship whatsoever** with any feature (including the target), removing those rows doesn't introduce bias — you're just working with a smaller, but still representative, sample.
- **Rule of thumb:** Use CCA only when missing data is a **small percentage** of the total (commonly cited threshold: **less than 5%**) across the relevant columns.

---

## Why It's Risky Beyond That Threshold

1. **Loses statistical power** — smaller sample size means less data to learn from, especially harmful for already-small datasets.

2. **Introduces bias if data isn't MCAR** — if missingness is related to another variable (**MAR** — Missing At Random) or related to the missing value itself (**MNAR** — Missing Not At Random), dropping those rows systematically removes a *non-random subset* of the population, skewing the remaining data's distribution.

   **Example:** If high-income respondents are more likely to leave the "income" field blank, dropping those rows biases your dataset toward *lower*-income cases — the CCA'd dataset no longer represents the true population.

3. **Compounds across multiple columns** — if several columns each have some missing values (even if each individually is under 5%), dropping any row with *any* missing value across *any* column can eliminate a much larger chunk of the dataset than expected.

   ```python
   print(f"Rows before: {len(df)}")
   print(f"Rows after CCA: {len(df.dropna())}")
   print(f"% data lost: {(1 - len(df.dropna())/len(df))*100:.2f}%")
   ```
   Always check this before committing to CCA — the loss can be surprisingly large.

---

So **CCA = "A: Remove them"** — just with its formal statistical name attached. It's the deletion-based approach that sits alongside (and is contrasted with) the various imputation techniques (B) and the Missing Indicator flagging approach you covered earlier.

---

# B. Impute Them

Instead of deleting, you **fill in** the missing values with estimated/substituted values. Split into Univariate (uses only the column itself) and Multivariate (uses other columns too) approaches.

## 1. Univariate Imputation

Only the missing column's own values are used to determine the fill value — no relationship with other features is considered.

### I. Numerical Data

**a. Mean or Median Imputation**

$$
x_{missing} = \bar{x} \quad \text{(mean)} \quad \text{or} \quad x_{missing} = \tilde{x} \quad \text{(median)}
$$

- **Mean:** Use when data is roughly symmetric/normally distributed, with few outliers.
- **Median:** Use when data is skewed or has outliers — median is robust to extreme values.
- **Downside:** Reduces variance of the feature (all missing values collapse to one value), and distorts the relationship between features (correlations get weakened) if a large fraction is missing.

**b. Random Value Imputation**

Fills each missing value with a value **randomly sampled from the existing (non-missing) distribution** of that same column.

- Preserves the original variance/distribution shape better than mean/median imputation, since you're not collapsing everything to a single point.
- Introduces randomness, so results can vary between runs (should set a random seed for reproducibility).
- Is applicable on both numerical and categorical data.

**c. End of Distribution Imputation**

Fills missing values with a value **far out in the tail** of the distribution (e.g., mean + 3×std, or the 99th percentile).

$$
x_{missing} = \bar{x} + 3\sigma
$$

- The idea: explicitly flags "this was missing" by pushing it to an extreme, distinguishable value — useful when missingness itself might carry information (i.e., **Missing Not At Random**, MNAR) and you want the model to be able to treat "missing" as its own distinct signal.
- Downside: can create artificial outliers that distort distance-based models (KNN, SVM) or skew scaling if applied after imputation.

### II. Categorical Data

**a. Mode Imputation (Most Frequent Value)**

Replace missing entries with the **most frequently occurring category** in that column.

- Simple, intuitive, works well when one category dominates and missingness is random.
- Downside: artificially inflates the frequency of the mode category, which can bias models that are sensitive to class balance.

**b. Missing as Its Own Category**

Instead of guessing a value, treat "missing" as a **valid category of its own** (e.g., replace NaN with the string `"Missing"`).

- Avoids making an assumption about what the true value might have been.
- Especially useful when missingness is **not random** — i.e., the fact that data is missing might itself be predictive (e.g., a survey field left blank might correlate with a certain type of respondent).

---

## 2. Multivariate Imputation

Uses **relationships with other features** in the dataset to estimate the missing value, rather than relying only on the column's own distribution.

### I. KNN Imputer

For each row with a missing value, it finds the **k nearest neighbor rows** (based on the other, non-missing features, using a distance metric like Euclidean distance) and fills the missing value using the **average (numerical)** or **mode (categorical)** of those neighbors.

$$
x_{missing} = \frac{1}{k}\sum_{i=1}^{k} x_i^{(neighbor)}
$$

- More accurate than univariate methods since it accounts for feature correlations — similar rows likely have similar values.
- **Important:** Since it relies on distance, features should be **scaled first** (e.g., StandardScaler), or features with larger ranges will dominate the distance calculation.
- Downside: computationally expensive for large datasets (distance to every row must be calculated), and sensitive to the choice of `k`.
- In sklearn: `KNNImputer(n_neighbors=k)`

### II. Iterative Imputer (Multivariate Imputation by Chained Equations — MICE)

Models each feature with missing values as a **function of all other features**, using regression, iteratively.

**How it works (simplified):**
1. Fill all missing values with a simple placeholder (e.g., mean) initially.
2. Pick one column with missing values; treat it as the target (`y`), and all other columns as predictors (`X`).
3. Train a regression model on rows where that column isn't missing, then predict/fill the missing values using the trained model.
4. Move to the next column with missing values, repeat step 2-3.
5. Repeat this cycle over multiple iterations (rounds) until the imputed values **converge** (stop changing significantly between iterations).

- Captures complex multivariate relationships between features far better than mean/median or even KNN.
- Computationally more expensive — trains multiple regression models across multiple iterations.
- In sklearn: `IterativeImputer` (in `sklearn.experimental`, still evolving as of recent versions).

---

## CCA vs. Imputation — Quick Comparison

| Aspect | CCA (Deletion) | Imputation |
|---|---|---|
| Data used | Only complete rows | All rows (missing values filled) |
| Assumption required | MCAR | Varies by method (some tolerate MAR/MNAR) |
| Information loss | High (entire rows dropped) | Low (values estimated, not discarded) |
| Bias risk | High if not MCAR | Lower, if imputation method matches missingness type |
| Simplicity | Very simple | Ranges from simple (mean) to complex (MICE) |
| Best for | Small % missing, MCAR confirmed | Larger % missing, or MAR/MNAR patterns |

---
# Missing Indicator

Rather than (or in addition to) imputing a value, you create a **new binary column** that flags whether a value was originally missing (1) or not (0) for each row.

$$
\text{indicator}_i = \begin{cases} 1, & x_i \text{ was missing} \\ 0, & x_i \text{ was not missing} \end{cases}
$$

- Useful when missingness itself is informative (MNAR) — e.g., a missing "income" field might correlate with unemployment, which is itself predictive of the target.
- Typically used **alongside** an imputation method (fill the value AND add the indicator column), so the model gets both a usable numeric value and the "this was originally missing" signal.
- In sklearn: `MissingIndicator`, or `SimpleImputer(add_indicator=True)` which adds it automatically.

---

# Summary Table

| Technique | Type | Uses Other Columns? | Best For |
|---|---|---|---|
| Remove (Deletion) | — | No | Small % missing, MCAR |
| Mean/Median | Univariate, Numerical | No | Symmetric data (mean) / skewed data (median) |
| Random Value | Univariate, Numerical | No | Preserving distribution shape |
| End of Distribution | Univariate, Numerical | No | Flagging MNAR as an extreme value |
| Mode | Univariate, Categorical | No | Dominant category, random missingness |
| Missing as Category | Univariate, Categorical | No | MNAR categorical data |
| KNN Imputer | Multivariate | ✅ Yes | Small-medium datasets, correlated features |
| Iterative Imputer (MICE) | Multivariate | ✅ Yes | Complex relationships, higher accuracy needs |
| Missing Indicator | Flagging technique | No (adds new column) | When missingness itself is predictive (MNAR) |

**Rule of thumb:** Start simple (mean/median/mode) as a baseline. If missingness seems non-random or the dataset has strong feature correlations, move to KNN or Iterative Imputation, and consider pairing any method with a Missing Indicator to preserve the "was this missing" signal.

# Machine Learning Pipelines

## What is a Pipeline?

A **Pipeline** (from `sklearn.pipeline`) is a way of **chaining multiple preprocessing steps and a final estimator (model) into a single object**, so that the entire workflow — transformations *and* model training/prediction — is executed in one sequential, reproducible unit.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

pipe = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('power_transform', PowerTransformer(method='yeo-johnson')),
    ('model', LogisticRegression())
])

pipe.fit(X_train, y_train)
predictions = pipe.predict(X_test)
```

Often combined with `ColumnTransformer` for mixed data types:

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), numeric_cols),
    ('cat', OneHotEncoder(), categorical_cols)
])

full_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('model', LogisticRegression())
])
```

## Why Pipelines Matter

1. **Prevents data leakage** — Each transformer's `fit()` (e.g., mean/std for scaling, λ for Box-Cox) is learned *only* on training data during `pipeline.fit()`, then correctly applied (`transform()` only, not re-fit) to test/validation data. Doing this manually is a common source of bugs where test data accidentally influences preprocessing.

2. **Reproducibility & cleanliness** — The entire workflow (transform → transform → model) becomes one object. No risk of forgetting a step, applying steps in the wrong order, or applying them inconsistently between training and inference.

3. **Simplifies cross-validation and hyperparameter tuning** — You can pass the whole pipeline into `GridSearchCV` or `cross_val_score`, and every fold correctly refits transformers on that fold's training data only:
```python
from sklearn.model_selection import GridSearchCV
params = {'model__C': [0.1, 1, 10]}
grid = GridSearchCV(full_pipeline, params, cv=5)
grid.fit(X_train, y_train)
```

4. **Production-readiness** — Once trained, the pipeline can be serialized (e.g., via `joblib` or `pickle`) as a single artifact, and deployed. New raw data just needs `pipeline.predict()` — no need to manually re-apply each preprocessing step at inference time.

5. **Cleaner, more maintainable code** — Instead of scattered preprocessing code, the whole flow is declared once, is easier to read, debug, and modify (e.g., swapping `StandardScaler` for `RobustScaler` is a one-line change).

**In short:** Pipelines turn a fragile, manually-ordered sequence of preprocessing + modeling steps into a single, leak-proof, reusable, and deployable object — which is why they're considered a best practice in any real-world ML workflow, not just a convenience.
