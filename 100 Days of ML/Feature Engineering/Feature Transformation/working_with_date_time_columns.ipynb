```python
import pandas as pd
import numpy as np
```

```python
date = pd.read_csv('orders.csv')
date.sample(5)
```

```python
time = pd.read_csv('messages.csv')
time.sample(5)
```

# Working with Date Column:

```python
date.sample(5)
```

```python
date.info()
```

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1: Extract Year:

```python
date['date_year'] = date['date'].dt.year
date.sample(5)
```

### 2: Extract Month:

```python
date['date_month'] = date['date'].dt.month
date.sample(5)
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.sample(5)
```

### 3: Extract Day:

```python
date['date_day'] = date['date'].dt.day
date.sample(5)
```

```python
date['date_day_of_week'] = date['date'].dt.dayofweek
date.sample(5)
```

```python
date['date_day_name'] = date['date'].dt.day_name()
date.sample(5)
```

```python
date['is_weekend'] = np.where(date['date_day_of_week'].isin([5, 6]), 1, 0)
date.sample(5)
```

```python
date['date_week_of_year'] = date['date'].dt.isocalendar().week
date.sample(5)
```

```python
date['date_quarter_of_year'] = date['date'].dt.quarter
date.sample(5)
```

```python
date['date_semester'] = np.where(date['date_quarter_of_year'].isin([1, 2]), 1, 2)
date.sample(5)
```

### Elapsed between Dates:

```python
import datetime

today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
(np.round((today - date['date']) / np.timedelta64(1, 'W'),0)) / 7
```

# Working with Time:

```python
time.head()
```

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
time.info()
```

```python
time['dates'] = time['date'].dt.date
time['time'] = time['date'].dt.time

time.head()
```

```python
time['hour'] = time['date'].dt.hour
time['minute'] = time['date'].dt.minute
time['second'] = time['date'].dt.second

time.head()
```

```python
today - time['date']
```

```python
(today - time['date']) / np.timedelta64(1, 'W')
```

```python
(today - time['date']) / np.timedelta64(1, 's')
```
