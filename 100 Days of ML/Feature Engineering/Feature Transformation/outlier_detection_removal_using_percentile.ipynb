# Outlier Detection Using the Percentile Method

**Category: Feature Transformation** (specifically, outlier handling — often paired with capping/trimming as part of preprocessing before scaling)

The percentile method identifies outliers by flagging any data point that falls **below or above a chosen percentile threshold** of the distribution — rather than relying on statistical measures like mean/std (Z-score) or IQR, it works directly off the data's rank/position.

---

## Core Idea

You pick a **lower percentile** and an **upper percentile** (commonly the 1st and 99th, or 5th and 95th), and treat anything falling outside that range as an outlier.

$$
\text{Lower bound} = P_{k}, \qquad \text{Upper bound} = P_{100-k}
$$

Where:
- $P_k$ = the value at the $k^{th}$ percentile of the data
- $k$ is typically a small number like 1 or 5

**Example (1st/99th percentile):**

$$
\text{Lower bound} = P_1, \qquad \text{Upper bound} = P_{99}
$$

Any value below $P_1$ or above $P_{99}$ is flagged as an outlier — meaning you're explicitly saying "the most extreme 1% on each end (2% total) of my data is considered anomalous."

---

## How It Differs from Other Outlier Methods

| Method | Basis | Sensitive to Distribution Shape? |
|---|---|---|
| **Z-score** | Mean & standard deviation | Assumes roughly normal/Gaussian data — distorted by outliers themselves (since they inflate mean/std) |
| **IQR (Interquartile Range)** | Q1, Q3, and 1.5×IQR rule | More robust, but fixed at Q1/Q3 (25th/75th percentile) — can't adjust how strict/lenient the cutoff is |
| **Percentile Method** | Any chosen percentile (1st/99th, 5th/95th, etc.) | Doesn't assume any distribution shape — works purely on rank/position, and you control exactly what % of data gets flagged |

The percentile method is essentially a **generalized, flexible version** of the IQR method — IQR is really just the percentile method with the cutoffs *fixed* at the 25th and 75th percentiles. With the percentile method, you choose the cutoffs yourself.

---

## Two Ways to Apply It

### 1. Trimming (Removal)

Simply **delete** rows where the feature value falls outside the chosen percentile range.

```python
lower = df['income'].quantile(0.01)
upper = df['income'].quantile(0.99)

df_trimmed = df[(df['income'] >= lower) & (df['income'] <= upper)]
```

- Simple, but **loses data** — same downside as Complete Case Analysis for missing values (loses information, shrinks dataset).
- Risky if outliers are legitimate rare events, not errors — you're discarding real signal.

### 2. Capping / Winsorization (Clipping)

Instead of removing rows, **replace** values beyond the bounds with the bound value itself — so extreme values get "pulled in" to the threshold rather than deleted.

$$
x' =
\begin{cases}
P_1, & x < P_1 \\
x, & P_1 \leq x \leq P_{99} \\
P_{99}, & x > P_{99}
\end{cases}
$$

```python
lower = df['income'].quantile(0.01)
upper = df['income'].quantile(0.99)

df['income'] = df['income'].clip(lower=lower, upper=upper)
# or equivalently:
df['income'] = np.where(df['income'] < lower, lower,
                 np.where(df['income'] > upper, upper, df['income']))
```

- **Preferred over trimming in most cases** — you keep all rows (no data loss), just compress the extreme values inward.
- This specific technique — capping using percentiles — is often called **Winsorization**.

---

## Choosing the Percentile Cutoff

There's no single "correct" percentile — it depends on how strict you want to be and how much of your data you're comfortable flagging:

| Cutoff | Data Flagged (both tails) | When to Use |
|---|---|---|
| 1st / 99th | 2% total | Conservative — only the most extreme values flagged |
| 5th / 95th | 10% total | More aggressive — wider net for what counts as an outlier |
| 0.5th / 99.5th | 1% total | Very conservative — for datasets with few, genuinely extreme errors |

**Practical approach:** visualize the distribution first (boxplot, histogram) to see how much of the tail actually looks anomalous before picking a cutoff — don't pick a percentile blindly.

---

## When to Use the Percentile Method (vs. IQR or Z-score)

- **Use Percentile Method when:** your data is **skewed or doesn't follow a normal distribution** (Z-score assumes normality, so it performs poorly on skewed data), and you want **precise control** over exactly what fraction of data gets treated as outliers — rather than being locked into IQR's fixed 25th/75th-based rule.
- **Downside:** it's somewhat arbitrary — the choice of "1%" or "5%" as cutoffs isn't derived from the data's actual statistical properties the way IQR's 1.5× multiplier or Z-score's ±3 threshold are; it's a judgment call you make upfront.

---

## Where This Fits in the Pipeline

Like other Feature Transformation steps, percentile-based capping should be **fit only on training data** (calculate `P1`/`P99` from `X_train`) and then applied using those same fitted bounds to `X_test` — never recalculate the percentiles on test data, or you'll leak information.

```python
from sklearn.preprocessing import FunctionTransformer

def percentile_cap(X, lower_q=0.01, upper_q=0.99):
    lower = np.quantile(X, lower_q, axis=0)
    upper = np.quantile(X, upper_q, axis=0)
    return np.clip(X, lower, upper)

cap_transformer = FunctionTransformer(percentile_cap)
```

*(Note: for a fully leak-proof, fit-on-train version usable inside a `Pipeline`/`GridSearchCV`, you'd typically wrap this in a custom transformer class with `fit()` storing the learned bounds and `transform()` applying them — `FunctionTransformer` alone recalculates on whatever data it's given each time, so it's fine for quick exploration but not strictly correct inside a CV loop.)*

## Capping (Winsorization) Technique:

````python
df_new = df.copy()

df_new['Height'] = np.where(
    df_new['Height'] >= upper_limit,
    upper_limit,
    np.where(
        df_new['Height'] <= lower_limit,
        lower_limit,
        df_new['Height']
    )
)
````
