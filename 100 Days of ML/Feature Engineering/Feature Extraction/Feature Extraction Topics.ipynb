# Feature Engineering:
The process of using domain knowledge to extract features from raw data. These features can be used to improve the performance of machine learning algorithms.

---

## Types:
- Feature Transformation
- Feature Construction
- Feature Selection
- Feature Extraction

---

# Feature Engineering: The Four Categories

Before diving in, here's the umbrella context — feature engineering is generally split into four categories:

| Category | Purpose |
|---|---|
| **Feature Transformation** | Change the *scale/distribution/representation* of existing features (values change, count of features usually stays the same) |
| **Feature Construction/Extraction** | *Create new* features from existing ones (e.g., PCA components, combining columns) |
| **Feature Selection** | *Choose a subset* of existing features, discarding the rest (no new values created) |
| **Feature Extraction** | Reduce dimensionality by deriving new compressed representations (e.g., PCA, LDA, autoencoders) |

---

# Feature Extraction

**Category: Feature Extraction is itself one of the four core categories of Feature Engineering** — distinct from Feature Transformation, Feature Construction, and Feature Selection.

Feature Extraction is the category directly motivated by the **Curse of Dimensionality** — its whole purpose is to compress a large number of (often correlated) features into a smaller number of dense, information-rich features.

---

## What Is Feature Extraction?

Feature Extraction is the process of **transforming the original feature space into a new, typically lower-dimensional space**, where the new features (often called "components" or "latent features") are mathematically derived combinations of the original ones — chosen to preserve as much of the meaningful structure/variance/information as possible.

Unlike Feature Construction (where *you* manually define a new feature like BMI using domain knowledge), Feature Extraction lets an **algorithm** discover the new representation — and the resulting features are usually **not directly interpretable** in real-world terms (e.g., "Principal Component 1" doesn't mean anything on its own the way "BMI" does).

---

## Why Feature Extraction Matters

1. **Fights the Curse of Dimensionality** — compresses many features into fewer dense ones, keeping the data denser and more tractable for distance-based and gradient-based algorithms.
2. **Removes redundancy/correlation** — when features are highly correlated, extraction techniques combine that shared information into fewer, uncorrelated components.
3. **Speeds up training** — fewer input dimensions means faster model fitting and lower memory use.
4. **Reduces overfitting risk** — fewer dimensions relative to sample size means less room for the model to fit noise.
5. **Enables visualization** — compressing to 2 or 3 dimensions lets you actually plot and visually inspect high-dimensional data.

---

## Common Techniques

### 1. PCA (Principal Component Analysis)

The most widely used extraction technique — covered in depth in your Dimensionality Reduction notebooks. Finds orthogonal directions (principal components) that capture maximum variance, ranked by importance, and projects the data onto the top few.

- **Type:** Unsupervised, linear
- Best for: general-purpose dimensionality reduction, removing multicollinearity, visualization

### 2. LDA (Linear Discriminant Analysis)

Similar mathematical machinery to PCA (eigenvectors/eigenvalues of a matrix), but **supervised** — instead of maximizing variance, it finds the directions that **best separate known class labels**.

$$
J(\mathbf{w}) = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}}
$$

Where $S_B$ is the between-class scatter matrix and $S_W$ is the within-class scatter matrix — LDA finds directions that maximize the ratio of between-class spread to within-class spread.

- **Type:** Supervised, linear
- Best for: classification tasks where you want dimensionality reduction that also improves class separability
- **Key difference from PCA:** PCA doesn't know about the target label; LDA explicitly uses it.

### 3. t-SNE (t-Distributed Stochastic Neighbor Embedding)

A **non-linear** technique that focuses on preserving *local* neighborhood structure — points that are close together in the original high-dimensional space stay close in the reduced (usually 2D/3D) space. Works by modeling pairwise similarities as probability distributions and minimizing the divergence between the high-dimensional and low-dimensional versions.

- **Type:** Unsupervised, non-linear
- Best for: **visualization only** (2D/3D plots of clusters) — not typically used as a preprocessing step before modeling, since it's computationally expensive and doesn't have a simple way to transform new/unseen points.
- **Caveat:** distances between clusters in a t-SNE plot aren't necessarily meaningful — only local neighborhood structure is preserved reliably.

### 4. UMAP (Uniform Manifold Approximation and Projection)

A newer alternative to t-SNE — also non-linear, also focused on preserving local (and somewhat more global) structure, but significantly faster and better at preserving overall data topology.

- **Type:** Unsupervised, non-linear
- Best for: visualization, and (unlike t-SNE) sometimes usable as an actual preprocessing step since it handles transforming new points more gracefully.

### 5. Autoencoders

A neural network architecture trained to reconstruct its own input, with a narrow "bottleneck" layer in the middle. The bottleneck's activations become the extracted, compressed features.

$$
\text{Input} \rightarrow \text{Encoder} \rightarrow \text{Bottleneck (compressed features)} \rightarrow \text{Decoder} \rightarrow \text{Reconstructed Output}
$$

- **Type:** Unsupervised, non-linear
- Best for: complex, high-dimensional data (images, text embeddings) where linear methods like PCA aren't expressive enough to capture the underlying structure.
- Trained by minimizing reconstruction error — the network is forced to learn a compressed representation that still contains enough information to rebuild the original input.

### 6. Kernel PCA

A variant of PCA that first projects data into a higher-dimensional space using a kernel function (similar to the "kernel trick" in SVMs), then applies standard PCA — allowing it to capture **non-linear** relationships that regular PCA would miss.

- **Type:** Unsupervised, non-linear
- Best for: data with non-linear structure where regular PCA's linear directions don't capture the true pattern.

## Summary Table

| Technique | Supervised? | Linear/Non-linear | Typical Use |
|---|---|---|---|
| PCA | No | Linear | General-purpose dimensionality reduction |
| LDA | Yes | Linear | Classification-oriented reduction |
| t-SNE | No | Non-linear | Visualization only (2D/3D) |
| UMAP | No | Non-linear | Visualization, sometimes preprocessing |
| Autoencoders | No | Non-linear | Complex data (images, embeddings) |
| Kernel PCA | No | Non-linear | Non-linear structure, still general-purpose |

## Feature Extraction vs. Feature Construction vs. Feature Selection

| Aspect | Feature Extraction | Feature Construction | Feature Selection |
|---|---|---|---|
| Driven by | Mathematical/statistical algorithms | Domain knowledge, manual logic | Statistical relevance / model importance |
| Effect on feature count | Decreases (compresses into fewer) | Increases (adds new alongside/instead of raw) | Decreases (drops a subset, keeps rest as-is) |
| Interpretability | Often low ("PC1" has no direct meaning) | High (you know exactly what "BMI" means) | High (features are unchanged, just fewer of them) |
| Creates new values? | Yes — mathematically derived combinations | Yes — logic-derived combinations | No — just keeps/drops existing columns |
| Examples | PCA, LDA, Autoencoders, t-SNE | Ratios, date decomposition, domain formulas | Filter/Wrapper/Embedded methods |
