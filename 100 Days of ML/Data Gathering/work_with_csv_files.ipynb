# Converters
Use to change the contents in a specific column like shortening the names etc

```python
pd.read_csv('D:/Important Files/Machine Learning/Datasets/IPL.csv').head(3)
```

```python
def rename(name):
    if name == 'Wankhede Stadium, Mumbai':
        return 'WS,M'
    if name == 'Brabourne Stadium, Mumbai':
        return 'BS,M'
    else:
        return name

pd.read_csv('D:/Important Files/Machine Learning/Datasets/IPL.csv',converters={'venue':rename}).head(5)
```
