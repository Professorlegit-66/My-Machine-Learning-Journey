```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
def analyzer(max_depth):
    data = pd.read_csv('D:/Important Files/Machine Learning/Datasets/Social_Network_Ads.csv')
    
    X = data[['Age', 'EstimatedSalary']].values
    y = data['Purchased'].values

    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(X, y)

    a = np.arange(start=X[:, 0].min() - 1, stop=X[:, 0].max() + 1, step=0.1)
    b = np.arange(start=X[:, 1].min() - 100, stop=X[:, 1].max() + 100, step=100)

    XX, YY = np.meshgrid(a, b)

    input_array = np.array([XX.ravel(), YY.ravel()]).T

    labels = clf.predict(input_array)

    plt.contourf(XX, YY, labels.reshape(XX.shape), alpha=0.5)
    plt.scatter(X[:, 0], X[:, 1], c=y)
    plt.xlabel('Age')
    plt.ylabel('EstimatedSalary')
    plt.show()
```

```python
analyzer(max_depth=None) #Overfitting
```

```python
analyzer(max_depth=1) #Underfitting
```

```python
analyzer(max_depth=2)
```

```python
analyzer(max_depth=5)
```

```python
analyzer(max_depth=10)
```

See the difference between max_depth=5 and max_depth=10. The model with max_depth=5 is less complex and may generalize better to unseen data, while the model with max_depth=10 may overfit the training data.
