```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1, n_redundant=0, n_classes=2,
                           n_clusters_per_class=1, random_state=41, hypercube=False, class_sep=15)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c = y, cmap='winter', s=100)
```

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
def perceptron(X, y):
	X = np.insert(X, 0, 1, axis=1)
	weights = np.ones(X.shape[1])
	lr = 0.1
	
	for i in range(1000):
		j = np.random.randint(0, X.shape[0])
		y_pred = step(np.dot(X[j], weights))
		weights = weights + lr * (y[j] - y_pred) * X[j]

	return weights[0], weights[1:]
```

```python
intercept_, coef_ = perceptron(X, y)
```

```python
print(f"Intercept: {intercept_}")
print(f"Coefficients: {coef_}")
```

```python
m = -(coef_[0] / coef_[1])
b = -(intercept_ / coef_[1])
```

```python
X_input = np.linspace(-3, 3, 100)
y_input = m * X_input + b
```

```python
plt.figure(figsize=(10, 6))
plt.plot(X_input, y_input, color='red', label='Decision Boundary', linewidth=2)
plt.scatter(X[:, 0], X[:, 1], c = 1-y, cmap='winter', s=100, label='Class 0')
plt.scatter(X[:, 0], X[:, 1], c = y, cmap='winter', s=100, label='Class 1')
plt.ylim(-3, 2)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Perceptron Decision Boundary')
plt.legend()
```

### Using SKLearn Logistic Regression:

The perceptron trick above finds *a* separating line, but it only guarantees convergence when classes are perfectly linearly separable, and even then the resulting boundary depends on the random sample order and epoch count rather than settling on an optimal solution. Since the perceptron update only updates on misclassified points, it stops as soon as it finds *any* boundary with zero errors — not necessarily the best one.

Scikit-learn's `LogisticRegression` is used here as a benchmark to compare against: it optimizes a proper loss function (log-loss) via a more robust solver, which converges to a well-defined optimum rather than an arbitrary separating line. Plotting both decision boundaries together shows how closely the from-scratch perceptron matches an established, production-grade implementation.

```python
lor = LogisticRegression()
lor.fit(X, y)
```

```python
print(f"Intercept: {lor.intercept_}")
print(f"Coefficients: {lor.coef_}")
```

```python
m_new = -(lor.coef_[0][0] / lor.coef_[0][1])
b_new = -(lor.intercept_ / lor.coef_[0][1])
```

```python
X_input1 = np.linspace(-3, 3, 100)
y_input1 = m_new * X_input1 + b_new
```

```python
plt.figure(figsize=(10, 6))
plt.plot(X_input, y_input, color='red', label='Perceptron Decision Boundary', linewidth=2)
plt.plot(X_input1, y_input1, color='black', label='Logistic Regression Decision Boundary', linewidth=2)
plt.scatter(X[:, 0], X[:, 1], c = 1-y, cmap='winter', s=100, label='Class 0')
plt.scatter(X[:, 0], X[:, 1], c = y, cmap='winter', s=100, label='Class 1')
plt.ylim(-3, 2)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Decision Boundaries')
plt.legend()
```
