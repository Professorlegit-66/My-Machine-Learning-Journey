```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import cross_val_score
from mlxtend.plotting import plot_decision_regions
```

```python
df = pd.read_csv('D:/Important Files/Machine Learning/Datasets/ushape.csv', header=None, names=['X', 'Y', 'class'])
df.head()
```

```python
X = df.iloc[:,0:2].values
y = df.iloc[:,-1].values
```

```python
plt.scatter(X[:,0], X[:,1], c=y, cmap='bwr')
```

### Plain Logistic Regression (Linear Decision Boundary):

```python
lr = LogisticRegression()
lr.fit(X, y)
```

```python
plot_decision_regions(X, y.astype('int'), clf=lr, legend=2)
```

```python
np.mean(cross_val_score(lr, X, y, scoring='accuracy', cv=5))
```

### Adding Polynomial Features:

```python
poly = PolynomialFeatures(degree=2)
x_trf = poly.fit_transform(X)
```

```python
lr2 = LogisticRegression()
lr2.fit(x_trf, y)
np.mean(cross_val_score(lr2, x_trf, y, scoring='accuracy', cv=5))
```

```python
def plot_decision_boundary(X,y,degree=1):
    
    poly = PolynomialFeatures(degree=degree)
    X_trf = poly.fit_transform(X)
    
    clf = LogisticRegression()
    clf.fit(X_trf,y)
    
    accuracy = np.mean(cross_val_score(clf,X_trf,y,scoring='accuracy',cv=10))
    
    a=np.arange(start=X[:,0].min()-1, stop=X[:,0].max()+1, step=0.01)
    b=np.arange(start=X[:,1].min()-1, stop=X[:,1].max()+1, step=0.01)


    XX,YY=np.meshgrid(a,b)
    
    input_array=np.array([XX.ravel(),YY.ravel()]).T

    labels=clf.predict(poly.transform(input_array))
    
    plt.contourf(XX,YY,labels.reshape(XX.shape),alpha=0.5)
    plt.scatter(X[:,0],X[:,1], c=y)
    plt.title('Degree = {}, accuracy is {}'.format(degree,np.round(accuracy,4)))
```

### Comparing Decision Boundaries Across Degrees:

```python
plot_decision_boundary(X, y)
```

```python
plot_decision_boundary(X, y, degree=2)
```

```python
plot_decision_boundary(X, y, degree=3)
```

```python
plot_decision_boundary(X, y, degree=5)
```

```python
plot_decision_boundary(X, y, degree=7)
```

## Observations:

- **Degree 1** (plain Logistic Regression) draws a straight line — clearly insufficient for U-shaped data.
- As degree increases, the boundary curves to match the U-shape, and cross-validated accuracy improve accordingly.
- At very high degrees (5, 7), watch for the boundary becoming overly wiggly or convoluted around individual points — a sign of overfitting, even if training accuracy still looks high. Cross-validation accuracy is the more reliable signal here, since it reflects performance on held-out folds rather than the exact training points.
