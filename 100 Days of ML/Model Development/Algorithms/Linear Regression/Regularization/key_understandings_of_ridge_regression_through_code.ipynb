```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from mlxtend.evaluate import bias_variance_decomp
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
```

```python
data = load_diabetes()
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['Target'] = data.target
```

```python
df.head()
```

```python
df.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=2)
```

### 1. How the Coefficients Get Affected:

```python
coefs = []
r2_scores = []

for i in [0, 10, 100, 1000]:
	rg = Ridge(alpha=i)
	rg.fit(X_train, y_train)

	coefs.append(rg.coef_.tolist())
	y_pred = rg.predict(X_test)
	r2_scores.append(r2_score(y_test, y_pred))
```

```python
plt.figure(figsize=(14,9))
plt.subplot(221)
plt.bar(data.feature_names, coefs[0])
plt.title('Alpha: 0 & R2-Score: {}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names, coefs[1])
plt.title('Alpha: 10 & R2-Score: {}'.format(round(r2_scores[1],2)))

plt.subplot(223)
plt.bar(data.feature_names, coefs[2])
plt.title('Alpha: 100 & R2-Score: {}'.format(round(r2_scores[2],2)))

plt.subplot(224)
plt.bar(data.feature_names, coefs[3])
plt.title('Alpha: 1000 & R2-Score: {}'.format(round(r2_scores[3],2)))
```

### 2. Higher Values Are Impacted More:

```python
alphas = [0, 0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000, 10000]
coef = []

for i in alphas:
	rg = Ridge(alpha=i)
	rg.fit(X_train, y_train)

	coef.append(rg.coef_.tolist())
```

```python
input_array = np.array(coef)
```

```python
coef_df = pd.DataFrame(input_array, columns=data.feature_names)
coef_df['Alphas'] = alphas
coef_df.set_index('Alphas')
```

```python
input_array = np.array(coef).T
```

```python
plt.figure(figsize=(15, 8))
plt.plot(alphas, np.zeros(len(alphas)), color='black', linewidth=2)

for i in range(input_array.shape[0]):
	plt.plot(alphas, input_array[i], label=data.feature_names[i])

plt.xscale('log')
plt.xlabel('Alpha')
plt.ylabel('Coefficient value')
plt.legend()
plt.show()
```

### 3. Impact on the Bias-Variance Tradeoff:

```python
m = 100
X = 5 * np.random.rand(m, 1) - 2
y = 0.7 * X**2 - 2 * X + 3 + np.random.randn(m, 1)

plt.scatter(X, y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X.reshape(100, 1), y.reshape(100), test_size=0.2, random_state=2)
```

```python
poly = PolynomialFeatures(degree=16)

X_train = poly.fit_transform(X_train)
X_test = poly.transform(X_test)
```

```python
alpha = np.linspace(0, 200, 100)

loss = []
bias = []
variance = []

for i in alpha:
	rg = Ridge(alpha=i)
	avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(rg, X_train, y_train, X_test, y_test, loss='mse', random_seed=123)

	loss.append(avg_expected_loss)
	bias.append(avg_bias)
	variance.append(avg_var)
```

```python
print(max(loss), max(bias), max(variance))
print(min(loss), min(bias), min(variance))
```

```python
plt.plot(alpha, loss, label='Loss')
plt.plot(alpha, bias, label='Bias')
plt.plot(alpha, variance, label='Variance')
plt.yscale('log')
plt.xlabel('Alpha')
plt.ylabel('Value (log scale)')
plt.legend()
```

### 4. Effect on the Loss Function:

```python
X, y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20, random_state=13)
```

```python
def cal_loss(m, alpha):
    return np.sum((y - m*X.ravel() + 2.29)**2) + alpha*m*m

def predict(m):
    return m*X - 2.29
```
