# Lasso Regression: Core Idea

Lasso (Least Absolute Shrinkage and Selection Operator) is linear regression with an added penalty term that discourages large coefficients. Instead of just minimizing prediction error, the model minimizes:

$$J(\theta) = \text{MSE}(\theta) + \alpha \sum_{i=1}^{n} |\theta_i|$$

The penalty is the **sum of absolute values** of the coefficients (the L1 norm), scaled by a hyperparameter α that controls how strong the penalty is.

---
# Key Understandings:

## 1. How the coefficients get affected

Each coefficient gets pulled toward zero during training. Unlike Ridge regression (L2 penalty), which shrinks coefficients smoothly and proportionally, Lasso can push coefficients to **exactly zero**. This happens because of the shape of the L1 penalty — it has a constant "pull" of magnitude α toward zero regardless of how large the coefficient currently is, whereas Ridge's pull grows weaker as the coefficient shrinks. That constant pull is strong enough to zero out weakly useful coefficients entirely, effectively removing those features from the model. This is why Lasso is used for **automatic feature selection**, not just shrinkage.

---

## 2. Higher values are impacted more (in terms of relative shrinkage) — actually the reverse

Worth clarifying this one, since it's a common mix-up: Lasso's absolute penalty term applies the *same fixed amount* of shrinkage to every coefficient, regardless of size. That means **small coefficients are affected proportionally more** than large ones — a coefficient of 0.05 shrunk by a fixed amount of, say, 0.03 loses 60% of its value, while a coefficient of 5 shrunk by the same 0.03 barely changes. This is exactly why small/weak coefficients tend to get driven to zero while genuinely important (large) coefficients survive relatively intact. This behavior is the mechanism behind sparsity.

---

## 3. Impact of bias-variance tradeoff

Increasing α increases bias and decreases variance:
- **Low α** → penalty is weak → model behaves close to plain linear regression → fits training data closely → low bias, high variance (risk of overfitting).
- **High α** → penalty dominates → more coefficients pushed to zero → simpler model → higher bias, lower variance (risk of underfitting).

The earlier plot is a direct illustration of this: at α=0.01 the curve wiggles to chase noise (low bias, high variance), while at α=1 it flattens into something closer to a straight line, missing the true curvature (high bias, low variance).

---

## 4. Effect on the loss function

The L1 penalty changes the geometry of the loss surface. Ridge's L2 penalty creates a smooth, differentiable bowl-shaped constraint region (a circle/sphere in coefficient space), so its minimum rarely lands exactly on an axis. Lasso's L1 penalty creates a constraint region with **sharp corners** (a diamond in 2D, a cross-polytope in higher dimensions). The optimal solution is more likely to land exactly on one of those corners, where one or more coefficients are exactly zero. This is also why the loss function becomes **non-differentiable at zero** — coordinate descent (rather than plain gradient descent) is typically used to optimize it, which is the solver seen in the earlier `ConvergenceWarning`.

---

## 5. Why it's called "Lasso"

The name is an acronym: **L**east **A**bsolute **S**hrinkage and **S**election **O**perator.
- **Shrinkage** — coefficients are shrunk toward zero.
- **Selection** — because some coefficients are shrunk *all the way* to zero, it performs implicit feature selection, keeping only a subset of predictors.

That combination — shrinking coefficients while simultaneously selecting a sparse subset of features — is Lasso's defining property and what distinguishes it from Ridge.
