## 5 Key Understandings:

### 1. How the Coefficients Get Affected

Ridge Regression adds a **penalty term** to the ordinary least squares (OLS) cost function, based on the squared magnitude of the coefficients:

$$J(w) = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{m} w_j^2$$

Because the objective now minimizes both prediction error *and* the size of the coefficients simultaneously, the coefficients get **shrunk toward zero** compared to what plain OLS would give — but never exactly to zero (unlike Lasso, which can zero them out completely). The larger $\lambda$ (the regularization strength) is set, the more aggressively every coefficient gets pulled down in magnitude.

This shrinkage isn't random — it specifically targets coefficients that are large or unstable, which leads to the next point.

---

### 2. Higher Values Are Impacted More

Since the penalty term is $w_j^2$ (squared), the penalty grows **quadratically** with coefficient size. A coefficient of 10 contributes a penalty of 100; a coefficient of 2 contributes only 4. This means:

- **Large coefficients get shrunk proportionally more** than small ones — Ridge disproportionately reins in whichever features had inflated, unstable weights (often caused by multicollinearity).
- Small coefficients are barely affected, since their contribution to the penalty is already tiny.

This is why Ridge is so effective against multicollinearity — when two features are highly correlated, OLS can assign one of them a huge positive coefficient and the other a huge negative one (since they cancel out predictively but blow up individually). Ridge's quadratic penalty punishes that instability directly and redistributes the weight more evenly between the correlated features.

---

### 3. Impact on the Bias-Variance Tradeoff

This is the core motivation for using Ridge at all. Plain OLS is **unbiased** but can have **high variance** — especially with multicollinearity or many features relative to data points, where small changes in the training data can swing coefficients wildly.

Ridge introduces a small amount of **bias** (coefficients are deliberately shrunk away from their "true" OLS values) in exchange for a **large reduction in variance** (the model becomes much more stable and less sensitive to which specific data points it happened to train on).

$$\text{Total Error} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

As long as the variance reduction outweighs the added bias, total error **decreases** — meaning the model generalizes better to unseen data, even though it fits the training data slightly worse than unregularized OLS would. This tradeoff is controlled entirely by $\lambda$:

| $\lambda$ | Bias | Variance |
|---|---|---|
| $\lambda = 0$ | Low (same as OLS) | High |
| $\lambda$ moderate | Slightly higher | Much lower |
| $\lambda \to \infty$ | Very high (coefficients → 0) | Very low |

The right $\lambda$ sits at the sweet spot minimizing total error — typically found via cross-validation.

---

### 4. Effect on the Loss Function

Plain OLS minimizes only the residual sum of squares:

$$J(w) = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

Ridge modifies this into a **constrained optimization problem** by adding the $L_2$ penalty term:

$$J(w) = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{m} w_j^2$$

Geometrically, this changes the shape of the loss surface — instead of one single minimum point (the true OLS solution), the loss surface now has an added cost for moving away from the origin in coefficient space. The overall minimum of this new function is pulled toward the origin, away from the unconstrained OLS minimum, in proportion to $\lambda$.

Since the added term is smooth and differentiable everywhere (unlike Lasso's $L_1$ penalty, which has a sharp corner at zero), Ridge Regression has a clean **closed-form solution**:

$$w = (X^TX + \lambda I)^{-1}X^Ty$$

Compare this to plain OLS's closed-form: $w = (X^TX)^{-1}X^Ty$. The only change is adding $\lambda I$ before inverting — this also has the side benefit of making the matrix always invertible, even when $X^TX$ itself is singular or near-singular (a common problem with severe multicollinearity).

---

### 5. Why It's Called "Ridge" Regression

The name comes directly from that $\lambda I$ term in the closed-form solution — adding $\lambda$ along the diagonal of the identity matrix. Visually and geometrically, this modification is like adding a "ridge" (a raised diagonal spine) to the matrix $X^TX$ before inverting it.

There's also a numerical-stability story behind the name: the technique was originally introduced (by Hoerl and Kennard in 1970) specifically to fix the problem of a **near-singular** $X^TX$ matrix (which happens under multicollinearity, since correlated columns make the matrix close to non-invertible). Adding $\lambda$ along the diagonal — the "ridge" — pushes the matrix's eigenvalues away from zero, stabilizing the inversion and guaranteeing a solution exists. The name reflects this literal structural addition to the matrix, not just a metaphor for the loss landscape.
