# Assumptions of Linear Regression:

## 1. Linear Relationship

The relationship between each independent variable ($X$) and the dependent variable ($y$) must actually be linear — i.e., $y$ can be expressed as a weighted sum of the $X$'s plus a constant.

- **Why it matters:** If the true relationship is curved (e.g., quadratic), a linear model will systematically under/over-predict in different regions.
- **How to check:** Scatter plots of each feature vs. target, or a plot of predicted vs. residual values (residuals should show no pattern).
- **Fix if violated:** Add polynomial/interaction terms, apply a transformation (log, sqrt), or switch to a non-linear model.

---

## 2. No Multicollinearity

Independent variables shouldn't be highly correlated with each other.

- **Why it matters:** When features are correlated, the model can't isolate each one's individual effect on $y$ — coefficient estimates become unstable and hard to interpret (small data changes can swing coefficients wildly, even flip their sign).
- **How to check:** Correlation matrix between features, or **VIF (Variance Inflation Factor)** — VIF > 5 or 10 usually flags a problem.
- **Fix if violated:** Drop one of the correlated features, combine them (e.g., PCA), or use regularization (Ridge handles this well since it shrinks correlated coefficients together).

---

## 3. Normality of Residuals

The **residuals** (errors: $y_i - \hat{y}_i$) should be approximately normally distributed. Note: this is about residuals, not the raw $X$ or $y$ data.

- **Why it matters:** Needed for valid confidence intervals, hypothesis tests, and p-values on coefficients. Point predictions themselves aren't affected — inference is.
- **How to check:** Q-Q plot of residuals, or a histogram of residuals, or statistical tests (Shapiro-Wilk).
- **Fix if violated:** Transform the target variable (log, Box-Cox), or just rely on point predictions if inference isn't your goal — with large samples, OLS estimates are still fine (Central Limit Theorem).

---

## 4. Homoscedasticity

The variance of the residuals should stay **constant** across all levels of the independent variable(s) — i.e., the spread of errors shouldn't grow or shrink as $X$ increases.

- **Why it matters:** If violated (**heteroscedasticity**), the model still gives unbiased coefficients, but standard errors are wrong — so confidence intervals and p-values become unreliable. Often shows up as a "cone" or "funnel" shape (common when the target has skewed magnitude, e.g., predicting prices where errors grow with price).
- **How to check:** Plot residuals vs. predicted values — should look like a random horizontal band, not a funnel/fan shape.
- **Fix if violated:** Log-transform the target, use weighted least squares, or use robust standard errors.

---

## 5. No Autocorrelation of Errors

Residuals should be independent of each other — the error at one point shouldn't help predict the error at another.

- **Why it matters:** Common violation in **time series data**, where consecutive errors tend to be correlated (today's error resembles yesterday's). This inflates the model's apparent confidence — standard errors get underestimated, making things look more statistically significant than they are.
- **How to check:** Durbin-Watson test (~2 means no autocorrelation), or plot residuals against time/order and look for patterns.
- **Fix if violated:** Add lagged variables, use time-series-specific models (ARIMA, etc.), or use Generalized Least Squares.

---

**Quick summary table:**

| Assumption | Checked via | If violated |
|---|---|---|
| Linearity | Residual vs. predicted plot | Add polynomial terms / transform |
| No multicollinearity | VIF, correlation matrix | Drop/combine features, regularize |
| Normal residuals | Q-Q plot, Shapiro-Wilk | Transform target |
| Homoscedasticity | Residual vs. predicted plot | Log-transform target, weighted LS |
| No autocorrelation | Durbin-Watson test | Add lags, time-series models |
