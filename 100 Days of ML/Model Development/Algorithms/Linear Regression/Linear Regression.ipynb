## Linear Regression

Linear regression models the relationship between input feature(s) and a continuous target as a **straight line (or flat plane, in higher dimensions)**: `y = m·x + b` for one feature, or `y = w₁x₁ + w₂x₂ + ... + b` for many. Training means finding the weights (`m`/`w`) and intercept (`b`) that minimize the sum of squared residuals — the vertical distances shown in the diagram.

## Types of Linear Regression

**1. Simple Linear Regression**
One input feature, one output. `y = mx + b`. The straight-line case — exactly what's in the diagram above. Used when you're modeling a target against a single predictor (e.g. house price vs. square footage).

**2. Multiple Linear Regression**
Multiple input features. `y = w₁x₁ + w₂x₂ + ... + wₙxₙ + b`. Instead of a 2D line, this fits a flat hyperplane through higher-dimensional space. Most real-world regression problems fall here — e.g. predicting house price from square footage, bedrooms, location, and age simultaneously.

**3. Polynomial Regression**
Still "linear" in the sense that it's linear in the *weights* (that's the technical definition of "linear" regression — not that the fitted curve is a straight line), but it fits curves by adding polynomial terms of the features: `y = w₁x + w₂x² + w₃x³ + ... + b`. Useful when the true relationship is curved, not straight — you're just feeding the model `x`, `x²`, `x³` etc. as additional "features," and ordinary linear regression fits weights to those.

**4. Regularized variants — Ridge, Lasso, Elastic Net**
These add a penalty term to the loss function to control overfitting, especially useful with many features or multicollinearity:
- **Ridge (L2)** — shrinks weights toward zero, but rarely to exactly zero. Good when many features contribute a little.
- **Lasso (L1)** — can shrink weights to *exactly* zero, effectively performing feature selection automatically. Good when you suspect only a few features actually matter.
- **Elastic Net** — combines both L1 and L2 penalties, useful when you want some feature selection but also stability with correlated features.
