```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.pipeline import Pipeline
```

```python
X = 6 * np.random.rand(200, 1) - 3
y = (0.8 * X**2) + (0.9 * X) + 2 + np.random.randn(200,1)
```

```python
plt.scatter(X, y, color='blue')
plt.xlabel('X')
plt.ylabel('y')
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
lr = LinearRegression()

lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
plt.scatter(X, y, color='blue')
plt.plot(X_test, lr.predict(X_test), color='red')
plt.xlabel('X')
plt.ylabel('y')
```

### Applying Polynomial Linear Regression:

```python
poly = PolynomialFeatures(degree=2)

X_train_trf = poly.fit_transform(X_train)
X_test_trf = poly.transform(X_test)
```

```python
print(X_train[0])
print(X_train_trf[0])
```

```python
lr_new = LinearRegression()

lr_new.fit(X_train_trf, y_train)
```

```python
y_pred_new = lr_new.predict(X_test_trf)
r2_score(y_test, y_pred_new)
```

```python
X_new = np.linspace(-3, 3, 200).reshape(200, 1)
X_new_poly = poly.transform(X_new)
y_new = lr_new.predict(X_new_poly)
```

```python
plt.plot(X_new, y_new, color='red', linewidth=2, label='Predictions')
plt.scatter(X_train, y_train, color='blue', label='Training Points')
plt.scatter(X_test, y_test, color='green', label='Testing Points')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
```

```python
def polynomial_regression(degree):
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)

    poly_features = PolynomialFeatures(degree=degree, include_bias=False)
    std_scaler = StandardScaler()
    lin_reg = LinearRegression()

    polynomial_regression = Pipeline([
        ('poly_features', poly_features),
        ('std_scaler', std_scaler),
        ('lin_reg', lin_reg)
    ])

    polynomial_regression.fit(X, y)
    y_new = polynomial_regression.predict(X_new)

    plt.plot(X_new, y_new, color='red', linewidth=2, label='Degree ' + str(degree))
    plt.scatter(X_train, y_train, color='blue', label='Training Points', linewidth=1)
    plt.scatter(X_test, y_test, color='green', label='Testing Points', linewidth=1)
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.axis([-3.5, 3.5, 0, 13])
    plt.legend(loc='upper left')
```

```python
polynomial_regression(20)
```

```python
polynomial_regression(60)
```
