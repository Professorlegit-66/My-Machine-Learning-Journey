```python
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
```

```python
x = 7 * np.random.rand(100, 1) - 2.8
y = 7 * np.random.rand(100, 1) - 2.8
z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y + 2 + np.random.randn(100, 1)
```

```python
df = px.data.iris()
fig = px.scatter_3d(df, x=x.ravel(), y=y.ravel(), z=z.ravel())
fig.show()
```

```python
X_combined = np.hstack((x, y))

X_train, X_test, z_train, z_test = train_test_split(X_combined, z, test_size=0.2, random_state=2)
```

```python
lr_linear = LinearRegression()
lr_linear.fit(X_train, z_train)

z_pred_lr = lr_linear.predict(X_test)
```

```python
x_input = np.linspace(x.min(), x.max(), 10)
y_input = np.linspace(y.min(), y.max(), 10)
xGrid, yGrid = np.meshgrid(x_input, y_input)
```

```python
final = np.vstack((xGrid.ravel().reshape(1, 100), yGrid.ravel().reshape(1, 100))).T
```

```python
z_final_lr = lr_linear.predict(final).reshape(10, 10)
```

```python
fig = px.scatter_3d(df, x=x.ravel(), y=y.ravel(), z=z.ravel())
fig.add_trace(go.Surface(x=x_input, y=y_input, z=z_final_lr))
fig.show()
```

### Using Polynomial Regression:

```python
poly = PolynomialFeatures(degree=2)

X_train_trf = poly.fit_transform(X_train)
X_test_trf = poly.transform(X_test)
```

```python
print('Input:', poly.n_features_in_)
print('Output:', poly.n_output_features_)
print('Powers:\n', poly.powers_)
```

```python
X_train_trf.shape
```

```python
lr_poly = LinearRegression()
lr_poly.fit(X_train_trf, z_train)

z_pred_poly = lr_poly.predict(X_test_trf)
```

```python
final_trf = poly.transform(final)
```

```python
z_final_poly = lr_poly.predict(final_trf).reshape(10, 10)
```

```python
fig = px.scatter_3d(df, x=x.ravel(), y=y.ravel(), z=z.ravel())
fig.add_trace(go.Surface(x=x_input, y=y_input, z=z_final_poly))
fig.update_layout(scene=dict(zaxis=dict(range=[0,35])))
fig.show()
```

### Comparing both R² Score:

```python
rmse_lr = np.sqrt(mean_squared_error(z_test, z_pred_lr))
rmse_poly = np.sqrt(mean_squared_error(z_test, z_pred_poly))

print(f"Linear Regression     — R²: {r2_score(z_test, z_pred_lr):.4f}, RMSE: {rmse_lr:.4f}")
print(f"Polynomial Regression — R²: {r2_score(z_test, z_pred_poly):.4f}, RMSE: {rmse_poly:.4f}")
```
