```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples=4, n_features=1, n_informative=1 ,n_targets=1, noise=80, random_state=13)
```

```python
plt.scatter(X,y)
```

```python
lr = LinearRegression()

lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='red')
```

## Gradient Descent:
- Equation is `y = mx + b`
- Assuming the slope (m) is 78.53 which would be constant throughout
- We take initial value of intercept (b) as 0

```python
y_pred = ((78.35 * X) + 0).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='blue', label = 'Ordinary Least Square (OLS)')
plt.plot(X, y_pred, color='red', label = 'b = 0')
plt.legend()
```

```python
# Interaion 1:
m = 78.53
b = 0

slope_at_b = -2 * np.sum(y - m*X.ravel() - b)
slope_at_b
```

```python
learning_rate = 0.1

step_size = learning_rate * slope_at_b
step_size
```

```python
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='blue', label = 'Ordinary Least Square (OLS)')
plt.plot(X, y_pred1, color='green', label = 'b = 20.91')
plt.plot(X, y_pred, color='red', label = 'b = 0')
plt.legend()
```

```python
# Interaion 2:

slope_at_b = -2 * np.sum(y - m*X.ravel() - b)
slope_at_b
```

```python
step_size = learning_rate * slope_at_b
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='blue', label = 'Ordinary Least Square (OLS)')
plt.plot(X, y_pred1, color='green', label = 'b = 20.91')
plt.plot(X, y_pred2, color='orange', label = 'b = 25.09')
plt.plot(X, y_pred, color='red', label = 'b = 0')
plt.legend()
```

```python
# Interaion 3:

slope_at_b = -2 * np.sum(y - m*X.ravel() - b)
slope_at_b
```

```python
step_size = learning_rate * slope_at_b
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='blue', label = 'Ordinary Least Square (OLS)')
plt.plot(X, y_pred1, color='green', label = 'b = 20.91')
plt.plot(X, y_pred2, color='orange', label = 'b = 25.09')
plt.plot(X, y_pred3, color='purple', label = 'b = 25.93')
plt.plot(X, y_pred, color='red', label = 'b = 0')
plt.legend()
```

```python
# Interaion 4:

slope_at_b = -2 * np.sum(y - m*X.ravel() - b)

step_size = learning_rate * slope_at_b

b = b - step_size
b
```

```python
y_pred4 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='blue', label = 'Ordinary Least Square (OLS)')
plt.plot(X, y_pred1, color='green', label = 'b = 20.91')
plt.plot(X, y_pred2, color='orange', label = 'b = 25.09')
plt.plot(X, y_pred3, color='purple', label = 'b = 25.93')
plt.plot(X, y_pred4, color='yellow', label = 'b = 26.09')
plt.plot(X, y_pred, color='red', label = 'b = 0')
plt.legend()
```

### Doing the same thing in loop:

```python
m = 78.35
b = -100
learning_rate = 0.1

epochs = 10

for i in range(epochs):
    slope_at_b = -2 * np.sum(y - m*X.ravel() - b)
    b = b - (learning_rate * slope_at_b)
    y_pred = (m * X) + b

    plt.plot(X, y_pred)

plt.scatter(X, y)
```

## Effect of smaller learning rate:

```python
m = 78.35
b = -100
learning_rate = 0.01

epochs = 10

for i in range(epochs):
    slope_at_b = -2 * np.sum(y - m*X.ravel() - b)
    b = b - (learning_rate * slope_at_b)
    y_pred = (m * X) + b

    plt.plot(X, y_pred)

plt.scatter(X, y)
```
