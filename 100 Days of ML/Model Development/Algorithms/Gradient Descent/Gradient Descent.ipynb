## What is Gradient Descent?

Gradient Descent is an **iterative optimization algorithm** used to find the values of parameters (like $m$ and $b$ in linear regression) that minimize a cost function. Instead of solving for the optimal parameters directly (like the closed-form OLS formula I used in `My_LR`), it starts with a guess and repeatedly nudges the parameters in the direction that reduces error, until it converges on the minimum.

Think of it like standing on a hill in thick fog and trying to reach the bottom valley — we can't see the whole landscape, but we can feel the slope under our feet at our current position, and we take a step in the steepest downhill direction. Repeat that enough times, and we reach the bottom.

---

## The Cost Function

For linear regression, the cost function is typically **Mean Squared Error**:

$$J(m, b) = \frac{1}{n}\sum_{i=1}^{n}(y_i - (mx_i + b))^2$$

This is a **convex, bowl-shaped** function when plotted against $m$ and $b$ — meaning it has exactly one minimum, no local minima to get stuck in (true for simple/multiple linear regression specifically; not true for many other models like neural networks).

Gradient Descent's job: find the $(m, b)$ pair sitting at the bottom of that bowl.

---

## The Core Idea: The Gradient

The **gradient** is the vector of partial derivatives of the cost function with respect to each parameter — it tells us the slope of the bowl at our current position, and points in the direction of **steepest increase**. Since we want to *decrease* cost, we move in the *opposite* direction of the gradient.

For MSE with respect to $m$ and $b$:

$$\frac{\partial J}{\partial m} = \frac{-2}{n}\sum_{i=1}^{n} x_i(y_i - (mx_i + b))$$

$$\frac{\partial J}{\partial b} = \frac{-2}{n}\sum_{i=1}^{n} (y_i - (mx_i + b))$$

---

## The Update Rule

At each iteration, update both parameters simultaneously:

$$m := m - \alpha \cdot \frac{\partial J}{\partial m}$$

$$b := b - \alpha \cdot \frac{\partial J}{\partial b}$$

Where $\alpha$ (alpha) is the **learning rate** — how big a step to take each iteration.

---

## The Algorithm, Step by Step

1. **Initialize** $m = 0$, $b = 0$ (or small random values)
2. **Compute predictions**: $\hat{y}_i = mx_i + b$ for all data points
3. **Compute gradients**: $\frac{\partial J}{\partial m}$ and $\frac{\partial J}{\partial b}$ using the formulas above
4. **Update parameters**: subtract $\alpha \times$ gradient from each
5. **Repeat** steps 2–4 for a fixed number of iterations (epochs) or until the cost stops decreasing meaningfully (convergence)

---

## The Learning Rate ($\alpha$) — Critical and Tricky

| Learning rate | What happens |
|---|---|
| Too small | Converges correctly but painfully slowly — wastes compute/time |
| Too large | Overshoots the minimum, may oscillate or even diverge (cost increases each step) |
| Just right | Smooth, efficient convergence to the minimum |

There's no universal "correct" value — it's typically tuned by trial and observing the cost curve (plot cost vs. iteration; it should decrease smoothly and flatten out).

---

## Why Use Gradient Descent Instead of the Closed-Form Solution?

My `My_LR` class uses the **Normal Equation** (closed-form OLS) — computing $m$ directly via the formula we wrote. That's exact and doesn't need a learning rate or iterations. So why does Gradient Descent exist?

| | Closed-form (Normal Equation) | Gradient Descent |
|---|---|---|
| Speed (small data, few features) | Fast | Slower (needs iterations) |
| Speed (large data, many features) | Slow — matrix inversion is $O(n^3)$ | Scales much better |
| Works for all cost functions? | No — only works when a closed-form solution exists | Yes — works for logistic regression, neural networks, anything differentiable |
| Needs tuning? | No | Yes (learning rate, iterations) |
| Guaranteed exact minimum? | Yes (in one step) | Approximate, depends on convergence |

This is exactly why Gradient Descent matters beyond linear regression — it's the optimization backbone for logistic regression, neural networks, and most of deep learning, where no closed-form solution exists at all.

---

## The Three Variants

**Batch Gradient Descent**
- Uses the **entire dataset** to compute the gradient at every single step.
- Very stable, smooth convergence — but slow per iteration on large datasets (recomputes over all $n$ rows every time).

**Stochastic Gradient Descent (SGD)**
- Uses **just one random data point** per iteration to estimate the gradient.
- Much faster per step, but noisy — the path to the minimum zig-zags rather than smoothly descending. Can actually help escape shallow local minima in non-convex problems (neural nets), though for simple linear regression that's not a concern since it's convex.

**Mini-Batch Gradient Descent**
- A middle ground: uses a small random **subset** (e.g. 32, 64 rows) per iteration.
- Balances SGD's speed with Batch's stability — this is what's actually used in practice for most real-world ML, especially deep learning.

---

## Visual Intuition

If we plotted cost $J$ against iterations for each variant:
- **Batch**: smooth, monotonically decreasing curve
- **SGD**: noisy curve that trends downward but jumps around
- **Mini-Batch**: somewhere in between — mostly smooth with some noise

This noise/smoothness tradeoff is exactly what we would want to visualize — plotting the cost curve for each variant side by side makes the difference concrete rather than just theoretical.
