```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import BaggingClassifier, RandomForestClassifier

from sklearn.datasets import make_classification
from sklearn.tree import plot_tree
```

```python
X, y = make_classification(n_features=5, n_redundant=0, n_informative=5, n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

## Bagging with `max_features=2`:

`BaggingClassifier` uses decision trees as its base model and trains 10 of them by default. Each tree gets a bootstrap sample of the rows, and because `max_features=2`
it also gets a random pair of the 5 columns, chosen **once per tree**. A tree can never use the other three columns.

```python
bag = BaggingClassifier(max_features=2)
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

### One bagged tree:

This plots the first tree. Its splits can only use two columns. Note that the labels `x[0]` and `x[1]` refer to the **tree's own two-column subset**,
not to `col1` and `col2` of the dataset, because each bagged tree is trained on just its subset.

```python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

## Random forest with `max_features=2`:

`RandomForestClassifier` trains 100 trees by default, each on a bootstrap sample of the rows. Here `max_features=2` means something different: at **every split**,
the tree considers only 2 randomly chosen features as candidates, and the draw is repeated at the next split. Over the whole tree, all 5 features can be used.

```python
rf = RandomForestClassifier(max_features=2)
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

### One forest tree:

This plots the fifth tree (`estimators_[4]`). It was trained on all five columns, so `x[0]` to `x[4]` match `col1` to `col5`.
Compare it with the bagged tree above: the splits should mix more different features, even though only 2 were considered at each split.

```python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```

## Summary:

- **Bagging with `max_features=2`:** the feature subset is picked once per tree, so each tree is limited to 2 of the 5 features.
- **Random forest with `max_features=2`:** the subset is re-drawn at every split, so each tree stays flexible while the trees still differ from each other.
- The two models here use different default tree counts (10 vs. 100), and there is no train/test split or random seed, so this notebook illustrates how the trees are *built*. It does not show which model is more accurate.
