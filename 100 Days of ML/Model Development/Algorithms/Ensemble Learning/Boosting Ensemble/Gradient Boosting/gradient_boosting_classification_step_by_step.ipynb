## Gradient Boosting for Classification: Step by Step:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor
from sklearn.tree import plot_tree

import warnings
warnings.filterwarnings('ignore', category=UserWarning)
```

```python
df = pd.DataFrame(
                  [
                      [6.82, 118, 0],
                      [6.36, 125, 1],
                      [5.39, 99, 1],
                      [5.50, 106, 1],
                      [6.39, 148, 0],
                      [9.13, 148, 1],
                      [7.17, 147, 1],
                      [7.72, 72, 0]
                   ], columns = ['cgpa', 'iq', 'is_placed']
)

df
```

## Stage 1: initial prediction:
5 of the 8 students were placed, so the starting log-odds is ln(5/3), the same prediction for every row.

```python
# initial prediction
df['pre1(log-odds)'] = np.log(5/3)
df
```

```python
# convert log odds to probability
df['pre1(probability)'] = 1 / (1 + np.exp(-np.log(5/3)))
df
```

**Residual = actual − predicted probability.** A positive residual means the model under-predicted placement.

```python
# calculating residual for stage 1
df['res1'] = df['is_placed'] - df['pre1(probability)']
df
```

## Stage 2: first tree:
Fit a small tree (3 leaves) to the stage-1 residuals, compute each leaf's log-odds update, and add it to the initial prediction to get `pre2`.

```python
# training the first decision tree
reg1 = DecisionTreeRegressor(max_leaf_nodes=3, random_state=1)
reg1.fit(df.iloc[:,0:2].values, df.iloc[:,-1].values)
```

```python
plot_tree(reg1, feature_names=['cgpa','iq'],filled=True, node_ids=True)
plt.show()
```

```python
df['leaf_entry1'] = reg1.apply(df.iloc[:,0:2])
df
```

### Leaf values:
Each leaf's log-odds update is Σ residuals / Σ p(1 − p) over the rows in that leaf.

```python
def return_logs(leaf):
  temp_df = df[df['leaf_entry1'] == leaf]
  num = temp_df['res1'].sum()

  den = sum(temp_df['pre1(probability)'] * (1 - temp_df['pre1(probability)']))
  return round(num/den,2)
```

```python
df['pre2(log-odds)'] = df['pre1(log-odds)'] + df['leaf_entry1'].apply(return_logs)
```

```python
df['pre2(probability)'] = 1 / (1 + np.exp(-df['pre2(log-odds)']))
df
```

```python
df['res2'] = df['is_placed'] - df['pre2(probability)']
df
```

## Stage 3: second tree:
Repeat on the stage-2 residuals: fit a new tree, compute its leaf values, and add them to `pre2` (not `pre1`, since `pre2` already contains it). Learning rate = 1, so no shrinkage.

```python
reg2 = DecisionTreeRegressor(max_leaf_nodes=3, random_state=1)
reg2.fit(df.iloc[:,0:2].values, df.iloc[:,-1].values)
```

```python
plot_tree(reg2, feature_names=['cgpa','iq'],filled=True, node_ids=True)
plt.show()
```

```python
df['leaf_entry2'] = reg2.apply(df.iloc[:,0:2])
df
```

```python
def return_logs(leaf):
  num = df[df['leaf_entry2'] == leaf]['res2'].sum()
  den = sum(df[df['leaf_entry2'] == leaf]['pre2(probability)'] * (1 - df[df['leaf_entry2'] == leaf]['pre2(probability)']))
  return round(num/den,2)
```

New log-odds = previous log-odds + this tree's leaf value (learning rate = 1, so no shrinkage).

```python
df['pre3(log-odds)'] = df['pre2(log-odds)'] + df['leaf_entry2'].apply(return_logs)
```

```python
df['pre3(probability)'] = 1 / (1 + np.exp(-df['pre3(log-odds)']))
df
```

```python
df['res_final'] = df['is_placed'] - df['pre3(probability)']
df[['res1','res2','res_final']]
```

The residuals generally get smaller at each stage, which is what boosting is doing: each tree corrects what the previous ones got wrong.
