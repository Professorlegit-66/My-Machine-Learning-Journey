# PCA Problem Formulation
PCA's "geometric intuition" tells us *what* we're looking for (directions of max variance) — the problem formulation is *how* we turn that into solvable math.

## Setting up the problem

**Step 1: Center the data**

Given a dataset with $n$ points and $d$ features, first subtract the mean from each feature so the data is centered at the origin:

$$X_{centered} = X - \bar{X}$$

This matters because variance calculations assume the data is centered — otherwise you'd be measuring spread relative to the wrong reference point.

**Step 2: Define what we're optimizing**

We want to find a unit vector $\vec{u}$ (a direction) such that when we project all data points onto it, the variance of those projections is maximized.

The projection of a data point $x_i$ onto direction $\vec{u}$ is $x_i \cdot \vec{u}$ (a scalar — how far along that direction the point lands).

So we want to maximize:

$$\text{Var}(X\vec{u}) = \frac{1}{n}\sum_{i=1}^{n} (x_i \cdot \vec{u})^2$$

**Step 3: Rewrite using the covariance matrix**

This variance expression can be rewritten in matrix form as:

$$\text{Var}(X\vec{u}) = \vec{u}^T \Sigma \vec{u}$$

where $\Sigma$ is the **covariance matrix** of the centered data — a $d \times d$ matrix capturing how every pair of features varies together.

**Step 4: Add the constraint**

Without a constraint, you could make $\vec{u}^T \Sigma \vec{u}$ arbitrarily large just by scaling $\vec{u}$ to be huge — that's meaningless. So we constrain $\vec{u}$ to be a **unit vector**:

$$\vec{u}^T \vec{u} = 1$$

This turns it into a constrained optimization problem:

$$\max_{\vec{u}} \ \vec{u}^T \Sigma \vec{u} \quad \text{subject to} \quad \vec{u}^T\vec{u} = 1$$

**Step 5: Solve with Lagrange multipliers**

Using a Lagrange multiplier $\lambda$, we form:

$$L(\vec{u}, \lambda) = \vec{u}^T \Sigma \vec{u} - \lambda(\vec{u}^T\vec{u} - 1)$$

Taking the derivative with respect to $\vec{u}$ and setting it to zero gives:

$$\Sigma \vec{u} = \lambda \vec{u}$$

**This is exactly the eigenvector equation.** It tells us that the direction $\vec{u}$ that maximizes variance must be an **eigenvector** of the covariance matrix $\Sigma$, and the variance captured along that direction equals its corresponding **eigenvalue** $\lambda$.

---

## Putting it together

- Compute the covariance matrix $\Sigma$ of the centered data.
- Find its eigenvectors and eigenvalues.
- Sort eigenvectors by their eigenvalues, largest first — the eigenvector with the largest eigenvalue is **PC1**, the next is **PC2**, and so on.
- Since $\Sigma$ is symmetric, its eigenvectors are guaranteed to be **orthogonal** — which is exactly why PC1, PC2, etc. come out perpendicular to each other, matching the geometric picture.

So the "problem formulation" step converts an intuitive geometric goal (find the direction of max spread) into a precise linear algebra problem (find eigenvectors of the covariance matrix) — which is also why PCA is efficiently computable even for high-dimensional data, since eigendecomposition is a well-understood, solvable operation.
