# PCA (Principal Component Analysis)
PCA (Principal Component Analysis) is fundamentally about finding a new set of axes to represent your data — axes chosen so that they capture the most "spread" (variance) in the data, in decreasing order of importance.

## The core idea

Imagine you have data with many features (dimensions). Geometrically, each data point is a point in that high-dimensional space. Most real datasets have correlated features — so when you plot them, the data doesn't fill the space uniformly; it tends to cluster along certain directions, like a cloud that's stretched out more in some directions than others.

PCA finds those "stretched-out" directions.

## Step by step, geometrically

1. **Center the data** — shift all points so the mean sits at the origin. This makes it easier to reason about directions purely as vectors from the center.

2. **Find the direction of maximum variance** — imagine spinning a line through the origin in every possible direction and asking: "if I project all my data points onto this line, how spread out do they become?" The direction where the projected points are most spread out is called the **first principal component (PC1)**.

3. **Find the next direction, orthogonal to the first** — among all directions perpendicular (orthogonal) to PC1, find the one that captures the next-most variance. That's **PC2**. This continues for as many dimensions as your data has.

4. **These new axes are uncorrelated** — because each principal component is orthogonal to the ones before it, the projected coordinates along these new axes have zero correlation with each other. PCA essentially "de-tangles" correlated features into independent directions.

5. **Dimensionality reduction** — since PC1, PC2, etc. are ranked by how much variance they capture, you can often keep just the first few components and discard the rest, losing minimal information while drastically cutting the number of dimensions.



## Why this matters
The whole point is that PC1 gives you the single best 1D summary of your data — if you had to compress everything down to one number per point, projecting onto PC1 preserves more information (variance) than any other possible line. PC2 gives the best *additional* direction, orthogonal to the first.

This is why PCA is used for dimensionality reduction: instead of describing each point using the original (possibly many, correlated) features, you describe it using its coordinates along PC1, PC2, PC3, etc. — fewer numbers, ranked by importance, and uncorrelated with each other.
