```python
import numpy as np
```

# Numpy Indexing and Slicing of Vectors

```python
arr = np.arange(1,11)
arr
```

```python
arr[6]
```

```python
arr[2:7]
```

```python
arr[4:]
```

```python
arr_slice = arr[::3]
arr_slice
```

# Numpy Indexing and Slicing of Matrix

```python
arr = np.arange(1,21).reshape(5,4)
arr
```

```python
arr[0]
```

```python
arr[0,0]
```

```python
arr[2,2]
```

```python
slice = arr[2:4,2:4]
slice
```

```python
slice = arr[3:,2:]
slice
```

```python
slice = arr[:,2]
slice
```

```python
slice = arr[2,:]
slice
```

# Boolean Indexing

```python
arr = np.arange(1,21)
arr
```

```python
bool_index = arr %2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
