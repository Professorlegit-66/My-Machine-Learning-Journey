# Array Operations
1: Arithmetic Operations

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
```

```python
sum = arr1 + arr2
sum
```

```python
sub = arr2 - arr1
sub
```

```python
mult = arr1 * arr2
mult
```

```python
arr1 / arr2
```

```python
arr1 // arr2
```

```python
arr1 ** arr2
```

# Advance Array Manipulation
1: Stacking Array

```python
mat1 = np.array([1,2,3,4])
mat2 = np.array([5,6,7,8])
```

```python
mat1
```

```python
mat2
```

```python
np.vstack((mat1,mat2))
```

```python
np.hstack((mat1,mat2))
```

```python
np.column_stack((mat1,mat2))
```

2: Splitting Array

```python
a = np.arange(16).reshape(4,4)
a
```

```python
np.hsplit(a,2)
```

```python
np.hsplit(a,4)
```

```python
np.vsplit(a,2)
```

```python
split = np.vsplit(a,4)
split
```

```python
for i in split:
    print(i)
```
