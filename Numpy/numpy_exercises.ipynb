# Valid Sudoku

```python
s = np.array([
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,8],
    [1,9,8,3,4,2,5,6,7],

    [8,5,9,7,6,1,4,2,3],
    [4,2,6,8,5,3,7,9,1],
    [7,1,3,9,2,4,8,5,6],

    [9,6,1,5,3,7,2,8,4],
    [2,8,7,4,1,9,6,3,5],
    [3,4,5,2,8,6,1,7,9]
])
s
```

```python
a = np.sum(s,axis = 1)
for i in a:
    if i != 45:
        print("Sudoku isn't valid")
        break
    else:
        print("For rows the Sudoku is valid")
        break
```

```python
b = np.sum(s,axis = 0)
for i in b:
    if i != 45:
        print("Sudoku isn't valid")
        break
    else:
        print("For Columns the Sudoku is valid")
        break
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3,j:j+3]
        print(n.sum())
```

# General Questions

```python
# Columns: Age, Math Marks, Science Marks
data = np.array([
    [18, 85, 78],
    [19, 92, 88],
    [17, 76, 95],
    [18, 65, 70],
    [20, 90, 85]
])
data
```

```python
# Get the shape of the matrix
data.shape
```

```python
# Find the average age of Students
slice = data[:,0]
slice
```

```python
sum = slice.sum()
average = sum / 5
average
```

```python
np.mean(data[:,0])
```

```python
#Extract the Math marks of all Students
slice = data[:,1]
slice
```

```python
#Find the highest Science marks
np.max(data[2])
```

```python
# Get details of the Students who scored higher than 90 in Math
data[data[:,1] > 90]
```

```python
# Increase the Math marks of the Students by 5
data[:,1] += 5
data
```

```python
# Find how many students are younger than 18
slice = data[:,0]
for i in slice:
    if i < 19:
        print("Student younger than 18 is",i)
```
