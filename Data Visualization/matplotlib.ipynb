```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# MatPlotLib Basics

```python
x = np.linspace(1,10,5)
y = x**2
```

```python
plt.plot(x,y)
plt.title('Basic Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
```

```python
plt.subplot(2,2,1)
plt.plot(x*1.5,y)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.plot(x*5,y)
plt.subplot(2,2,3)
plt.plot(y/2,x)
plt.subplot(2,2,4)
plt.plot(y,x)
```

# Object Oriented Way

```python
fig = plt.figure(figsize=(10,6), dpi=100)
axis1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.3, 0.25, 0.25])
axis2.plot(x, y)

plt.show()
```

# Types of Plots

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(x)
```

# Saving Plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.plot(x,y)
plt.savefig('Basic Plot.png')
```
