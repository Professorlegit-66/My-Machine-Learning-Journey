```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# Categorical Data Plots

```python
df = sns.load_dataset('tips')
df.head()
```

1: Count Plot

```python
sns.countplot(x = df['sex'], hue = df['smoker'])
```

2: Bar Plot

```python
sns.barplot(x = df['sex'],y = df['tip'])
```

```python
import numpy as np

sns.barplot(x = df['sex'],y = df['total_bill'], estimator= np.sum)
```

3: Box Plot

```python
sns.boxplot(x=df['tip'], y=df['day'], palette='rainbow')
```

4: Violin Plot

```python
sns.violinplot(x=df['day'], y=df['total_bill'], palette='rainbow')
```

5: Strip Plot

```python
sns.stripplot(x=df['day'], y= df['tip'], palette='rainbow')
```

6: Swarm Plot

```python
sns.swarmplot(x= df['day'], y=df['total_bill'], palette='rainbow')
```

```python
sns.violinplot(x= df['day'], y=df['total_bill'], palette='rainbow')
sns.swarmplot(x= df['day'], y=df['total_bill'], palette='rainbow')
```
