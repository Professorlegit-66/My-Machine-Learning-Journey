1: Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(x = match_wins.values, y = match_wins.index, palette='viridis')
plt.title('Most matches winner')
```

2: Toss decision trends

```python
sns.countplot(x=df['toss_decision'], palette='viridis')
plt.title('Toss decision trends')
```

3: Toss winner vs Match winner

```python
count = df[df['toss_winner'] == df['match_winner']]['match_winner'].count()
percentage = (count * 100) / df.shape[0]
percentage.round(2)
```

# Key Player Performance

```python
df.head()
```

1: Most "Player of the match" awards

```python
awards = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x=awards.values, y=awards.index)
```

2: Two Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

3: Top 10 best bowlling figures

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
bowlling = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
bowlling.plot(kind='barh')
```
