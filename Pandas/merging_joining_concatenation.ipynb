```python
import numpy as np
import pandas as pd
```

# Merging 2 Databases

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name': ['John','Anna','Peter','Linda','Bob'],
    'department': ['HR','IT','Finance','IT','HR']
})

salaries = pd.DataFrame({
    'employee_id': [1,2,3,6,7],
    'salaries': [60000, 62000, 65000, 70000, 76000],
    'bonus': [5000, 10000, 8000, 12000, 7000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner')
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
```

# Concatenation of 2 Databases

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], axis=1)
```

# Joining of 2 Databases

```python
Df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie']
}, index= [1,2,3])

Df2 = pd.DataFrame({
    'Score': [85, 90, 75]
}, index=[2,3,4])
```

```python
Df1
```

```python
Df2
```

```python
Df1.join(Df2)
```

```python
Df1.join(Df2,how='outer')
```

```python
Df1.join(Df2,how='right')
```
