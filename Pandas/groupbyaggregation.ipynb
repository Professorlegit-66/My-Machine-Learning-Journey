```python
import numpy as np
import pandas as pd
```

# Group By

```python
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 130, 200, 300],
    'Quantity': [10, 15, 12, 20, 8, 9, 15, 25],
    'Data': pd.date_range('2026-07-18', periods=8)
})
df
```

```python
df.groupby('Category')
```

```python
df.groupby('Category')['Sales'].sum()
```

```python
group = df.groupby('Store')
for i, v in group:
    print(i)
    print(v)
```

```python
df.groupby('Store')['Sales'].sum()
```

```python
df.groupby(['Category', 'Store'])['Sales'].sum()
```

```python
df.groupby(['Store', 'Category'])['Sales'].sum()
```
