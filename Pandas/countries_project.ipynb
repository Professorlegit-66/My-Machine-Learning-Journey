```python
# Which Country has the highest population

df[df['population'] == df['population'].max()]
```

```python
# What is the capital of the highest population country

df[df['population'] == df['population'].max()]['capital_city']
```

```python
# Which Country has the least population

df[df['population'] == df['population'].min()]['country']
```

```python
# What is the capital of the least population country

df[df['population'] == df['population'].min()]['capital_city']
```

```python
# Give me top 5 Countries with the highest democratic score

df['democracy_score'].nlargest(5)
```

```python
df.sort_values(by = 'democracy_score', ascending=False,inplace=True)
df.head(5)
```

```python
# How many total regions are there

df['region'].value_counts()
```

```python
df['region'].value_counts().count()
```

```python
# How many Countries lie in Eastern Europe region

df[df['region'] == 'Eastern Europe']['country']
```

```python
# Who is the political leader of the second highest population Country

df[df['population'] == df['population'].nlargest(2).iloc[1]]['political_leader']
```

```python
# How many countries are there whose political leaders are unknown

df[df['political_leader'].isna()]['country'].count()
```

```python
# How many countries have Republic in their names

count = 0

def find_republic(txt):
    global count
    if 'republic' in txt.lower():
        count += 1
    return txt

df['country_long'].apply(find_republic)
print(count)
```

```python
# Which Country in African region has highest population

african_df = df[df['continent'] == 'Africa']
african_df
```
